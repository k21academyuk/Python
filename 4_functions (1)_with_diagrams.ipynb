# Part 1: What is a Function?

A function is like a **machine**: you give it some input, it does a job, and it gives back a result.

```
                     +------------+
  arguments  ----->  |  function  |  ----->  return value
  (input)            +------------+           (output)
```

You have already used many functions: `print()`, `len()`, `input()`, `type()`.

**Why do we need functions?**

| Benefit | Meaning |
|---|---|
| **Reuse** | Write the code once, call it many times |
| **Less repetition** | No copy-pasting the same lines (the DRY rule: Don't Repeat Yourself) |
| **Readability** | A good name like `calculate_tax()` explains what the code does |
| **Easy to fix** | Fix a bug in one place and every call gets the fix |
| **Teamwork** | Different people can write different functions |

**Kinds of functions in Python**

| Kind | Example |
|---|---|
| Built-in functions | `len()`, `max()`, `sorted()` |
| Functions from modules | `math.sqrt()`, `random.randint()` |
| User-defined functions | functions you create with `def` |
| Lambda (anonymous) functions | `lambda x: x * 2` |
| Generator functions | functions that use `yield` |

````python
# Example: Without a function, the same lines are repeated
print('Hello, Asha!')
print('Welcome to the Python course.')
print('Hello, Ravi!')
print('Welcome to the Python course.')

# With a function: write the lines once, reuse them (Part 3 explains def)
def welcome(name):
    print(f'Hello, {name}!')
    print('Welcome to the Python course.')

welcome('Meena')
welcome('John')
# Expected output:
# Hello, Asha!
# Welcome to the Python course.
# Hello, Ravi!
# Welcome to the Python course.
# Hello, Meena!
# Welcome to the Python course.
# Hello, John!
# Welcome to the Python course.
````

# Part 2: Built-in Functions

**Built-in functions** are ready-made functions that come with Python. You can use them anywhere without importing anything. Python has about 70 of them; these are the ones you will use most:

| Group | Functions | What they do |
|---|---|---|
| Input and output | `print()`, `input()` | Show output, read keyboard input |
| Type conversion | `int()`, `float()`, `str()`, `bool()`, `list()`, `tuple()`, `set()`, `dict()` | Convert a value to another type |
| Maths | `abs()`, `round()`, `min()`, `max()`, `sum()`, `pow()`, `divmod()` | Common calculations |
| Sequences | `len()`, `sorted()`, `reversed()`, `range()`, `enumerate()`, `zip()` | Work with strings, lists and other collections |
| Checking | `type()`, `isinstance()`, `all()`, `any()` | Ask questions about values |
| Characters | `ord()`, `chr()` | Convert between characters and their code numbers |
| Help | `help()`, `dir()` | Find out what something does |

````python
# Example 1: Maths functions
print(abs(-7))            # absolute value (distance from 0)
print(round(3.14159, 2))  # round to 2 decimal places
print(round(7.6))         # round to the nearest whole number
print(pow(2, 5))          # 2 to the power 5, same as 2 ** 5
print(divmod(17, 5))      # (quotient, remainder) together
# Expected output:
# 7
# 3.14
# 8
# 32
# (3, 2)
````

````python
# Example 2: Functions that work on a whole collection
marks = [78, 92, 65, 88]
print(len(marks))
print(min(marks), max(marks))
print(sum(marks))
print(sum(marks) / len(marks))
# Expected output:
# 4
# 65 92
# 323
# 80.75
````

````python
# Example 3: sorted() returns a NEW sorted list; the original is unchanged
scores = [40, 10, 30, 20]
print(sorted(scores))
print(sorted(scores, reverse=True))
print(list(reversed(scores)))
print(scores)
# Expected output:
# [10, 20, 30, 40]
# [40, 30, 20, 10]
# [20, 30, 10, 40]
# [40, 10, 30, 20]
````

````python
# Example 4: Type conversion functions
print(int('25') + 5)
print(float('2.5') * 2)
print(str(100) + ' rupees')
print(list('abc'))
print(set([1, 2, 2, 3]))
# Expected output:
# 30
# 5.0
# 100 rupees
# ['a', 'b', 'c']
# {1, 2, 3}
````

````python
# Example 5: Checking functions
print(type(3.5))
print(isinstance(10, int))
print(all([True, True, False]))   # True only if ALL items are True
print(any([False, False, True]))  # True if ANY item is True
print(ord('A'), chr(66))
# Expected output:
# <class 'float'>
# True
# False
# True
# 65 B
````

````python
# Example 6: enumerate() and zip()
names = ['Asha', 'Ravi']
scores = [91, 84]
for i, name in enumerate(names, start=1):
    print(i, name)
for name, score in zip(names, scores):
    print(name, 'scored', score)
# Expected output:
# 1 Asha
# 2 Ravi
# Asha scored 91
# Ravi scored 84
````

````python
# Example 7: help() shows the documentation of any function
help(abs)
# Expected output (similar to):
# Help on built-in function abs in module builtins:
#
# abs(x, /)
#     Return the absolute value of the argument.
````

# Part 3: Defining Your Own Functions (User-Defined Functions)

You create your own function with the `def` keyword.

**Syntax:**
```python
def function_name(parameters):
    """Docstring: what the function does (optional)."""
    statements
    return value          # optional
```

| Part | Meaning |
|---|---|
| `def` | Keyword that starts a function definition |
| `function_name` | The name used to call it. Same rules as variable names; use lowercase words joined by `_`, usually starting with a verb (`get_total`, `print_report`) |
| `(parameters)` | Names for the input values. Can be empty: `()` |
| `:` | The colon ends the header line |
| Indented body | The code that runs when the function is called (indented 4 spaces) |
| `return` | Sends a result back to the caller (optional) |

**Defining vs calling**
- **Defining** (`def ...`) only creates the function. Nothing inside it runs yet.
- **Calling** (`function_name()`) runs the body. You can call a function as many times as you like.

**Diagram: Anatomy of a function**

Anatomy of a function

````python
# Example 1: A function with no parameters, called twice
def say_hello():
    print('Hello!')
    print('Have a great day.')

print('Before calling')
say_hello()      # call 1
say_hello()      # call 2
print('After calling')
# Expected output:
# Before calling
# Hello!
# Have a great day.
# Hello!
# Have a great day.
# After calling
````

````python
# Example 2: Defining a function does NOT run it
def show_message():
    print('You will only see this when the function is called')

print('The function is defined, but not called yet')
# Expected output: The function is defined, but not called yet
````

````python
# Example 3: Forgetting the parentheses does not call the function
def say_hi():
    print('Hi!')

print(say_hi)    # shows the function object itself, not 'Hi!'
say_hi()         # the () is what runs it
# Expected output:
# <function say_hi at 0x...>   (the number changes every time)
# Hi!
````

````python
# Example 4: A function must be defined BEFORE it is called
# In a fresh session, calling it first gives: NameError: name 'later' is not defined
# later()          # (uncomment to see the error in a fresh session)

def later():
    print('Now it works')

later()
# Expected output: Now it works
````

````python
# Example 5: An empty function body needs pass
def calculate_tax():
    pass     # TODO: write this later

calculate_tax()   # runs and does nothing
print('No error')
# Expected output: No error
````

````python
# Example 6: A function body can contain any code: loops, if statements, other calls
def print_even_numbers():
    for n in range(1, 11):
        if n % 2 == 0:
            print(n, end=' ')
    print()

print_even_numbers()
# Expected output: 2 4 6 8 10
````

# Part 4: Parameters and Arguments

A function becomes much more useful when you can give it **input**.

| Term | Where it appears | Example |
|---|---|---|
| **Parameter** | In the function **definition**: a variable name waiting for a value | `def greet(name):` here `name` is a parameter |
| **Argument** | In the function **call**: the actual value you pass in | `greet('Asha')` here `'Asha'` is an argument |

When the function is called, each argument is **assigned** to its parameter (like `name = 'Asha'`), and then the body runs.

**Rule:** by default, the number of arguments must match the number of parameters.

````python
# Example 1: One parameter
def greet(name):
    print(f'Good morning, {name}!')

greet('Asha')
greet('Ravi')
# Expected output:
# Good morning, Asha!
# Good morning, Ravi!
````

````python
# Example 2: Two parameters - arguments are matched by position
def show_area(length, width):
    print(f'Area of a {length} x {width} rectangle is {length * width}')

show_area(5, 3)
show_area(10, 2)
# Expected output:
# Area of a 5 x 3 rectangle is 15
# Area of a 10 x 2 rectangle is 20
````

````python
# Example 3: Arguments can be values, variables or expressions
price = 250
show_area(price // 50, 2 + 2)
# Expected output: Area of a 5 x 4 rectangle is 20
````

````python
# Example 4: The wrong number of arguments gives a TypeError
# show_area(5)        # TypeError: show_area() missing 1 required positional argument: 'width'
# show_area(5, 3, 1)  # TypeError: show_area() takes 2 positional arguments but 3 were given
print('Every call to show_area() must give exactly 2 arguments')
# Expected output: Every call to show_area() must give exactly 2 arguments
````

````python
# Example 5: Passing user input as an argument
def print_table(number):
    for i in range(1, 6):
        print(f'{number} x {i} = {number * i}')

n = int(input('Enter a number: '))
print_table(n)
# Example run:
# Enter a number: 4
# 4 x 1 = 4
# 4 x 2 = 8
# 4 x 3 = 12
# 4 x 4 = 16
# 4 x 5 = 20
````

### Passing Lists and Dictionaries

When you pass a list or a dictionary, the function receives the **same object**, not a copy. Changes made inside the function (like `append`) are visible outside it.

Numbers and strings cannot be changed in place. Giving the parameter a new value inside the function does not change the variable outside.

````python
# Example 6: A function can change a list that is passed in
def add_item(shopping_list, item):
    shopping_list.append(item)

my_list = ['milk']
add_item(my_list, 'bread')
add_item(my_list, 'eggs')
print(my_list)
# Expected output: ['milk', 'bread', 'eggs']
````

````python
# Example 7: Reassigning a number inside a function does not change the outside variable
def try_to_change(x):
    x = x + 100
    print('Inside the function:', x)

value = 5
try_to_change(value)
print('Outside the function:', value)
# Expected output:
# Inside the function: 105
# Outside the function: 5
````

# Part 5: Return Values

`return` sends a result **back to the caller**, so it can be stored in a variable, printed, or used in a calculation.

**print vs return**

| `print()` | `return` |
|---|---|
| Shows a value on the screen | Hands a value back to the code that called the function |
| The value cannot be used later | The value can be stored, compared or calculated with |
| A function can print many times | `return` **ends the function immediately** |

A function without `return` (or with a bare `return`) gives back the special value `None`.

**Diagram: How a function call and return work**

How a function call and return work

````python
# Example 1: Returning a value and storing it
def square(n):
    return n * n

result = square(6)
print(result)
print(square(3) + square(4))     # return values used in an expression
# Expected output:
# 36
# 25
````

````python
# Example 2: print vs return
def add_print(a, b):
    print(a + b)

def add_return(a, b):
    return a + b

x = add_print(2, 3)     # shows 5, but gives back None
y = add_return(2, 3)    # shows nothing, gives back 5
print('x =', x)
print('y =', y)
print('y * 10 =', y * 10)
# Expected output:
# 5
# x = None
# y = 5
# y * 10 = 50
````

````python
# Example 3: return ends the function immediately (early return)
def check_age(age):
    if age < 0:
        return 'Invalid age'
    if age >= 18:
        return 'Adult'
    return 'Minor'
    print('This line never runs')

print(check_age(25))
print(check_age(12))
print(check_age(-3))
# Expected output:
# Adult
# Minor
# Invalid age
````

````python
# Example 4: Returning True or False - handy inside if statements
def is_even(n):
    return n % 2 == 0

for number in [4, 7]:
    if is_even(number):
        print(number, 'is even')
    else:
        print(number, 'is odd')
# Expected output:
# 4 is even
# 7 is odd
````

````python
# Example 5: Returning more than one value (Python packs them into a tuple)
def min_max_avg(numbers):
    return min(numbers), max(numbers), sum(numbers) / len(numbers)

stats = min_max_avg([10, 20, 30, 40])
print(stats)
low, high, average = min_max_avg([10, 20, 30, 40])   # unpack into 3 variables
print('Lowest:', low, 'Highest:', high, 'Average:', average)
# Expected output:
# (10, 40, 25.0)
# Lowest: 10 Highest: 40 Average: 25.0
````

````python
# Example 6: A function with no return gives back None
def say(text):
    print(text)

answer = say('Hello')
print(answer)
print(type(answer))
# Expected output:
# Hello
# None
# <class 'NoneType'>
````

````python
# Example 7: Functions can call other functions
def get_grade(marks):
    if marks >= 90:
        return 'A'
    elif marks >= 75:
        return 'B'
    elif marks >= 50:
        return 'C'
    return 'F'

def report(name, marks):
    grade = get_grade(marks)          # calling another function
    return f'{name}: {marks} marks, grade {grade}'

print(report('Asha', 92))
print(report('Ravi', 67))
# Expected output:
# Asha: 92 marks, grade A
# Ravi: 67 marks, grade C
````

# Part 6: Types of Arguments

| Type | How it works | Example call |
|---|---|---|
| **Positional** | Matched to parameters by **position** (order) | `describe('dog', 'Tommy')` |
| **Keyword** | Matched by **parameter name**, so order does not matter | `describe(name='Tommy', animal='dog')` |
| **Default** | The parameter has a value in the definition, so the argument can be **left out** | `def power(base, exponent=2):` |

**Rules**
1. In a call, positional arguments must come **before** keyword arguments.
2. In a definition, parameters with defaults must come **after** parameters without defaults.
3. You cannot give the same parameter two values.

**Diagram: Positional, keyword and default arguments**

Positional, keyword and default arguments

````python
# Example 1: Positional arguments - order matters
def describe(animal, name):
    print(f'I have a {animal} named {name}.')

describe('dog', 'Tommy')
describe('Tommy', 'dog')     # wrong order gives a silly result
# Expected output:
# I have a dog named Tommy.
# I have a Tommy named dog.
````

````python
# Example 2: Keyword arguments - order does not matter
describe(name='Tommy', animal='dog')
describe(animal='cat', name='Kitty')
# Expected output:
# I have a dog named Tommy.
# I have a cat named Kitty.
````

````python
# Example 3: Mixing positional and keyword arguments
describe('parrot', name='Mithu')       # positional first, then keyword: OK
# describe(animal='parrot', 'Mithu')   # SyntaxError: positional argument follows keyword argument
# Expected output: I have a parrot named Mithu.
````

````python
# Example 4: Default parameter values
def power(base, exponent=2):
    return base ** exponent

print(power(5))                  # exponent uses its default, 2
print(power(2, 10))              # the default is replaced by 10
print(power(base=3, exponent=3))
# Expected output:
# 25
# 1024
# 27
````

````python
# Example 5: Several default values - give only the ones you want to change
def make_coffee(size='medium', sugar=1, milk=True):
    milk_text = 'with milk' if milk else 'black'
    print(f'A {size} coffee, {sugar} sugar, {milk_text}')

make_coffee()
make_coffee('large')
make_coffee(sugar=0, milk=False)
# Expected output:
# A medium coffee, 1 sugar, with milk
# A large coffee, 1 sugar, with milk
# A medium coffee, 0 sugar, black
````

````python
# Example 6: You already used keyword arguments with defaults: print(sep=' ', end=new line)
print('a', 'b', 'c')
print('a', 'b', 'c', sep='-')
print('Loading', end='...')
print('done')
# Expected output:
# a b c
# a-b-c
# Loading...done
````

````python
# Example 7: Rule - parameters with defaults must come last
# def wrong(a=1, b):    # SyntaxError: a parameter without a default cannot follow one with a default
def right(b, a=1):
    return a + b

print(right(5))
# Expected output: 6
````

### Common Trap: A List as a Default Value

Default values are created **once**, when the function is defined, not every time it is called. If the default is a list, the **same list** is shared by every call.

**Fix:** use `None` as the default and create a new list inside the function.

````python
# Example 8: The shared default list problem and its fix
def add_score_wrong(score, scores=[]):
    scores.append(score)
    return scores

print(add_score_wrong(10))
print(add_score_wrong(20))   # surprise: 10 is still there!

def add_score(score, scores=None):
    if scores is None:
        scores = []          # a brand-new list on every call
    scores.append(score)
    return scores

print(add_score(10))
print(add_score(20))
# Expected output:
# [10]
# [10, 20]
# [10]
# [20]
````

# Part 7: Variable-Length Arguments - `*args` and `**kwargs`

Sometimes you do not know in advance how many arguments will be passed.

| Syntax | Collects | Into a | Example call |
|---|---|---|---|
| `*args` | any number of extra **positional** arguments | **tuple** | `total(1, 2, 3)` |
| `**kwargs` | any number of extra **keyword** arguments | **dictionary** | `show_profile(name='Asha', city='Delhi')` |

The names `args` and `kwargs` are only a convention; the `*` and `**` are what matter.

**Order in a definition:** regular parameters first, then `*args`, then `**kwargs`:
```python
def order_pizza(size, *toppings, **extras):
```

The stars also work the other way round in a **call**: `*my_list` spreads a list into separate positional arguments, and `**my_dict` spreads a dictionary into keyword arguments.

**Diagram: How *args and **kwargs collect values**

How *args and **kwargs collect values

````python
# Example 1: *args collects extra positional arguments into a tuple
def total(*numbers):
    print(numbers)
    return sum(numbers)

print(total(1, 2))
print(total(5, 10, 15, 20))
print(total())
# Expected output:
# (1, 2)
# 3
# (5, 10, 15, 20)
# 50
# ()
# 0
````

````python
# Example 2: A regular parameter followed by *args
def print_scores(student, *scores):
    print(f'{student} took {len(scores)} tests')
    for s in scores:
        print('  score:', s)

print_scores('Asha', 78, 85, 91)
# Expected output:
# Asha took 3 tests
#   score: 78
#   score: 85
#   score: 91
````

````python
# Example 3: **kwargs collects extra keyword arguments into a dictionary
def show_profile(**details):
    print(details)
    for key, value in details.items():
        print(f'{key}: {value}')

show_profile(name='Ravi', age=21, city='Delhi')
# Expected output:
# {'name': 'Ravi', 'age': 21, 'city': 'Delhi'}
# name: Ravi
# age: 21
# city: Delhi
````

````python
# Example 4: A regular parameter, *args and **kwargs together
def order_pizza(size, *toppings, **extras):
    print(f'Size: {size}')
    print('Toppings:', toppings)
    print('Extras:', extras)

order_pizza('large', 'onion', 'corn', cheese='double', delivery=True)
# Expected output:
# Size: large
# Toppings: ('onion', 'corn')
# Extras: {'cheese': 'double', 'delivery': True}
````

````python
# Example 5: Unpacking a list or dictionary when CALLING a function
def describe(animal, name):   # same function as in Part 6
    print(f'I have a {animal} named {name}.')

pet = ['rabbit', 'Bunny']
pet_info = {'animal': 'fish', 'name': 'Goldie'}
describe(*pet)         # same as describe('rabbit', 'Bunny')
describe(**pet_info)   # same as describe(animal='fish', name='Goldie')
# Expected output:
# I have a rabbit named Bunny.
# I have a fish named Goldie.
````

````python
# Example 6: print() itself uses *args - that is why it accepts any number of values
print('one')
print('one', 'two', 'three', 4, 5.0)
numbers = [1, 2, 3]
print(*numbers)          # unpacks the list into separate arguments
# Expected output:
# one
# one two three 4 5.0
# 1 2 3
````

# Part 8: Variable Scope - Local, Global and LEGB

**Scope** means *where* a variable can be used.

| Scope | Created | Can be used |
|---|---|---|
| **Local** | inside a function | only inside that function; it disappears when the function ends |
| **Global** | outside all functions (top level of the notebook or file) | everywhere, but a function can only **read** it unless it says `global` |

### The LEGB Rule
When Python meets a name, it searches these places in order and uses the **first** match:
1. **L**ocal: inside the current function
2. **E**nclosing: inside any outer function (for a function defined inside another function)
3. **G**lobal: the top level of the file
4. **B**uilt-in: Python's own names, like `len` and `print`

**Diagram: The LEGB scope rule**

The LEGB scope rule

````python
# Example 1: A local variable exists only inside its function
def make_greeting():
    message = 'Hello from inside'    # local variable
    print(message)

make_greeting()
# print(message)   # NameError: name 'message' is not defined
# Expected output: Hello from inside
````

````python
# Example 2: Functions can READ global variables
school = 'Green Valley School'     # global variable

def show_school():
    print('I study at', school)

show_school()
# Expected output: I study at Green Valley School
````

````python
# Example 3: A local variable with the same name hides the global one
count = 100            # global

def counter():
    count = 5          # a NEW local variable; the global is untouched
    print('Inside:', count)

counter()
print('Outside:', count)
# Expected output:
# Inside: 5
# Outside: 100
````

````python
# Example 4: Changing a global variable needs the global keyword
visits = 0

def add_visit():
    global visits      # without this line: UnboundLocalError
    visits += 1

add_visit()
add_visit()
add_visit()
print('Visits:', visits)
# Expected output: Visits: 3
````

````python
# Example 5: The cleaner way - pass the value in and return the new value
def add_visit_clean(current):
    return current + 1

visits = 0
visits = add_visit_clean(visits)
visits = add_visit_clean(visits)
print('Visits:', visits)
# Expected output: Visits: 2
````

````python
# Example 6: Enclosing scope - a nested function can read its outer function's variables
def outer():
    greeting = 'Hi'              # enclosing variable for inner()
    def inner():
        name = 'Asha'            # local to inner()
        print(greeting, name)    # greeting is found in the Enclosing scope
    inner()

outer()
# Expected output: Hi Asha
````

````python
# Example 7: Built-in scope - and why you should never reuse built-in names
print(len('hello'))      # len is found in the Built-in scope

sum = 10                 # BAD: this global name now hides the built-in sum()
# print(sum([1, 2]))     # TypeError: 'int' object is not callable
del sum                  # delete our variable so the built-in sum() works again
print(sum([1, 2]))
# Expected output:
# 5
# 3
````

# Part 9: Documenting Functions - Docstrings and Type Hints

**Docstring:** a string in triple quotes written as the **first line** of the function body. It explains what the function does, its parameters and what it returns. `help()` displays it.

**Type hints:** optional labels showing the expected type of each parameter and of the return value, such as `price: float` and `-> float`. They make code easier to read, but **Python does not check or enforce them**.

````python
# Example 1: Writing a docstring and reading it with help()
def rectangle_area(length, width):
    """Return the area of a rectangle.

    length: the length of the rectangle
    width: the width of the rectangle
    """
    return length * width

help(rectangle_area)
# Expected output (similar to):
# Help on function rectangle_area in module __main__:
#
# rectangle_area(length, width)
#     Return the area of a rectangle.
#
#     length: the length of the rectangle
#     width: the width of the rectangle
````

````python
# Example 2: Type hints describe the expected types
def final_price(price: float, discount_percent: float = 0) -> float:
    """Return the price after the discount."""
    return price - price * discount_percent / 100

print(final_price(500, 10))
print(final_price(200))
# Expected output:
# 450.0
# 200.0
````

````python
# Example 3: Type hints are NOT enforced - they are only labels
def double(n: int) -> int:
    return n * 2

print(double(4))
print(double('ab'))   # no error, even though 'ab' is not an int
# Expected output:
# 8
# abab
````

# Part 10: Functions are Objects

In Python a function is a value, just like a number or a list. That means you can:
- store a function in another variable
- put functions inside a list or a dictionary
- **pass a function as an argument** to another function

A function that takes another function as an argument is called a **higher-order function**. `sorted(..., key=...)`, `map()` and `filter()` are built-in higher-order functions.

````python
# Example 1: Giving a function a second name
def shout(text):
    return text.upper() + '!'

yell = shout          # no (): we are not calling it, just naming it again
print(yell('hello'))
# Expected output: HELLO!
````

````python
# Example 2: Passing a function as an argument
def apply_twice(func, value):
    return func(func(value))

def add_ten(n):
    return n + 10

print(apply_twice(add_ten, 5))
print(apply_twice(shout, 'hi'))
# Expected output:
# 25
# HI!!
````

````python
# Example 3: A dictionary of functions - a tiny calculator
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

operations = {'+': add, '-': subtract, '*': multiply}
for symbol in ['+', '-', '*']:
    print(8, symbol, 2, '=', operations[symbol](8, 2))
# Expected output:
# 8 + 2 = 10
# 8 - 2 = 6
# 8 * 2 = 16
````

````python
# Example 4: sorted() with key - sort words by their length
def word_length(word):
    return len(word)

words = ['banana', 'kiwi', 'apple', 'fig']
print(sorted(words))                    # alphabetical
print(sorted(words, key=word_length))   # by length, shortest first
# Expected output:
# ['apple', 'banana', 'fig', 'kiwi']
# ['fig', 'kiwi', 'apple', 'banana']
````

# Part 11: Lambda Functions

A **lambda** is a small, one-line function **without a name**.

**Syntax:**
```python
lambda parameters: expression
```
The result of the expression is returned automatically; there is no `return` keyword.

**The same function written both ways:**
```python
def square(x):          # normal function
    return x * x

lambda x: x * x         # the same logic as a lambda
```

**When to use a lambda:** for a short function that is needed only once, usually as an argument to `sorted()`, `map()` or `filter()`.

**Limitations of lambda**
- Only **one expression**: no loops, no `if` statements (only the one-line `a if condition else b`), no multiple lines, no assignments
- No name and no docstring, so long or reused logic should be a normal `def` function
- Becomes hard to read as soon as it grows long

````python
# Example 1: Simple lambdas
square = lambda x: x * x            # given a name here only for this demo
plus = lambda a, b: a + b
print(square(7))
print(plus(3, 4))
print((lambda name: 'Hi ' + name)('Asha'))   # created and called at once
# Expected output:
# 49
# 7
# Hi Asha
````

````python
# Example 2: lambda as the key for sorted()
students = [('Asha', 91), ('Ravi', 78), ('Meena', 85)]
by_marks = sorted(students, key=lambda s: s[1], reverse=True)
print(by_marks)
by_length = sorted(['banana', 'kiwi', 'apple'], key=lambda w: len(w))
print(by_length)
# Expected output:
# [('Asha', 91), ('Meena', 85), ('Ravi', 78)]
# ['kiwi', 'apple', 'banana']
````

````python
# Example 3: map() applies a function to every item
prices = [100, 250, 400]
with_tax = list(map(lambda p: p * 1.05, prices))
print(with_tax)
names = ['asha', 'ravi']
print(list(map(str.title, names)))     # map() also works with normal functions
# Expected output:
# [105.0, 262.5, 420.0]
# ['Asha', 'Ravi']
````

````python
# Example 4: filter() keeps only the items for which the function returns True
marks = [35, 78, 42, 90, 28]
passed = list(filter(lambda m: m >= 40, marks))
print(passed)
evens = list(filter(lambda n: n % 2 == 0, range(1, 11)))
print(evens)
# Expected output:
# [78, 42, 90]
# [2, 4, 6, 8, 10]
````

````python
# Example 5: A lambda can use a one-line if-else, but not statements
result = lambda m: 'Pass' if m >= 40 else 'Fail'
print(result(55), result(30))
# bad = lambda x: for i in range(x): print(i)   # SyntaxError: loops are not allowed in a lambda
# Expected output: Pass Fail
````

# Part 12: Recursion

A **recursive function** is a function that **calls itself** to solve a smaller version of the same problem.

Every recursive function needs two parts:

| Part | Meaning | factorial example |
|---|---|---|
| **Base case** | The simplest case, answered directly **without** calling itself. This is what stops the recursion | `factorial(1)` is `1` |
| **Recursive case** | The function calls itself with a **smaller** input, moving towards the base case | `factorial(n) = n * factorial(n - 1)` |

Without a base case the function calls itself forever, and Python stops it with a **RecursionError** (the default limit is about 1000 nested calls).

**Diagram: How factorial(3) runs**

How factorial(3) runs

````python
# Example 1: A recursive countdown
def countdown(n):
    if n == 0:                 # base case
        print('Go!')
        return
    print(n)
    countdown(n - 1)           # recursive case: a smaller problem

countdown(3)
# Expected output:
# 3
# 2
# 1
# Go!
````

````python
# Example 2: Factorial with recursion (5! = 5 x 4 x 3 x 2 x 1)
def factorial(n):
    if n <= 1:                 # base case
        return 1
    return n * factorial(n - 1)

print(factorial(3))
print(factorial(5))
# Expected output:
# 6
# 120
````

````python
# Example 3: Sum of digits with recursion
def digit_sum(n):
    if n < 10:                          # base case: a single digit
        return n
    return n % 10 + digit_sum(n // 10)  # last digit + sum of the other digits

print(digit_sum(4567))
# Expected output: 22
````

````python
# Example 4: The same factorial with a loop
def factorial_loop(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

print(factorial_loop(5))
# Expected output: 120
````

### Recursion or Loop?

- Every recursive function can also be written with a loop (Example 4 above).
- Recursion suits problems that contain smaller copies of themselves: folders inside folders, family trees, some maths formulas.
- For simple counting, a loop is usually simpler and faster.

**Never forget the base case.** This function has none, so it keeps calling itself until Python stops it:
```python
def forever(n):
    return forever(n - 1)

forever(5)   # RecursionError: maximum recursion depth exceeded
```

# Part 13: Generator Functions

A **generator function** looks like a normal function, but uses `yield` instead of `return`.

| `return` | `yield` |
|---|---|
| Sends back one result and **ends** the function | Sends back one value and **pauses** the function |
| Calling the function runs its body | Calling the function only creates a **generator object**; the body runs when you ask for values |
| Produces everything at once | Produces values **one at a time, on demand** |

You get values from a generator with:
- `next(gen)`: runs the body until the next `yield` and gives that value
- a `for` loop: calls `next()` for you until the generator is finished

**Why use generators?** They save memory: each value is created only when it is needed, instead of storing a whole list.

**Diagram: How a generator pauses and resumes**

How a generator pauses and resumes

````python
# Example 1: A simple generator
def three_numbers():
    yield 1
    yield 2
    yield 3

gen = three_numbers()      # nothing runs yet; we just get a generator object
print(gen)
print(next(gen))
print(next(gen))
print(next(gen))
# Expected output:
# <generator object three_numbers at 0x...>   (the number changes)
# 1
# 2
# 3
````

````python
# Example 2: Watch the generator pause at each yield and resume at the next request
def steps():
    print('  Step A runs')
    yield 'first'
    print('  Step B runs')
    yield 'second'
    print('  Generator finished')

g = steps()
print('Got:', next(g))
print('Got:', next(g))
# next(g)   # a third call prints '  Generator finished', then raises StopIteration
# Expected output:
#   Step A runs
# Got: first
#   Step B runs
# Got: second
````

````python
# Example 3: A generator with a loop inside, used in a for loop
def count_up_to(limit):
    n = 1
    while n <= limit:
        yield n
        n += 1

for number in count_up_to(5):
    print(number, end=' ')
print()
# Expected output: 1 2 3 4 5
````

````python
# Example 4: Collecting a generator's values with list() and sum()
def even_numbers(limit):
    for n in range(2, limit + 1, 2):
        yield n

print(list(even_numbers(10)))
print(sum(even_numbers(100)))
# Expected output:
# [2, 4, 6, 8, 10]
# 2550
````

````python
# Example 5: A generator can be used only once
gen = even_numbers(6)
print(list(gen))
print(list(gen))     # already used up, so it is empty
# Expected output:
# [2, 4, 6]
# []
````

````python
# Example 6: A practical generator - ticket numbers produced on demand
def ticket_numbers(prefix):
    number = 101
    while True:              # endless, but safe: it only runs when asked for a value
        yield f'{prefix}-{number}'
        number += 1

tickets = ticket_numbers('BUS')
for _ in range(3):
    print(next(tickets))
# Expected output:
# BUS-101
# BUS-102
# BUS-103
````

# Part 14: Quick Recap

| Concept | Example |
|---|---|
| Define a function | `def greet(name):` |
| Call a function | `greet('Asha')` |
| Return a value | `return a + b` |
| Return several values | `return low, high` |
| Default argument | `def power(base, exponent=2):` |
| Keyword argument | `power(exponent=3, base=2)` |
| Any number of positional arguments | `def total(*numbers):` |
| Any number of keyword arguments | `def show_profile(**details):` |
| Change a global variable | `global visits` |
| Docstring | `"""What the function does."""` |
| Lambda | `lambda x: x * 2` |
| Recursion | a function that calls itself, with a base case |
| Generator | `yield value`, read with `next()` or a `for` loop |

### Common beginner mistakes
1. **Forgetting `()`** when calling: `say_hi` instead of `say_hi()`.
2. **Calling a function before defining it**, which gives a `NameError`.
3. **Using `print` when you need `return`**: the caller gets `None`.
4. **Writing code after `return`**: it never runs.
5. **Wrong number or order of arguments**: `TypeError`, or silly results.
6. **Using a list as a default value** (`scores=[]`): use `None` instead.
7. **Changing a global variable without `global`**: `UnboundLocalError`. Better: pass it in and return the new value.
8. **Reusing built-in names** like `sum`, `list` or `max` as variable names.
9. **Recursion without a base case**: `RecursionError`.
10. **Reusing a generator that is already used up**: it gives nothing.

# Part 16: Final Project - Smart Grocery Billing System

Time to combine **everything from all four labs** into one small, real-world program: a billing system for a grocery shop.

**What the program does**

2. Adds items to a customer's cart, checking for wrong codes and quantities
3. Calculates the subtotal, a discount based on the bill amount, and tax
4. Prints a neat bill with the most expensive items first
5. Runs as a menu until the customer checks out

**Where each concept is used**

| Lab | Concepts | Used in |
|---|---|---|
| Basic Syntax | variables, constants, f-strings, `input()`, typecasting | prices, `TAX_PERCENT`, the menu |
| Data Structures | dictionary, list, tuple, set | `PRODUCTS`, the cart, `DISCOUNT_RULES`, categories |
| Control Flow | `if-elif-else`, `for`, `while`, `break`, `continue` | checking input, the menu loop |
| Functions | parameters, default and keyword arguments, `return`, several return values | `add_to_cart()`, `calculate_bill()` |
| Functions | `*args`, `**kwargs`, `lambda`, generator | `add_many()`, `print_bill()`, sorting, `line_items()` |

Build it step by step, and **run every step in order**.

**Diagram: How the project functions connect**

How the project functions connect

### Step 1: Store the Data

The products are stored in a **dictionary of dictionaries**. Names in UPPER_CASE are constants: values that should not change while the program runs. The discount rules are a **list of tuples**, and a **set** collects each category only once.

````python
# Step 1: Product catalogue, tax and discount rules
PRODUCTS = {
    'P1': {'name': 'Rice (1 kg)', 'price': 60.0, 'category': 'Grains'},
    'P2': {'name': 'Milk (1 L)', 'price': 45.0, 'category': 'Dairy'},
    'P3': {'name': 'Bread', 'price': 40.0, 'category': 'Bakery'},
    'P4': {'name': 'Eggs (12)', 'price': 72.0, 'category': 'Dairy'},
    'P5': {'name': 'Apples (1 kg)', 'price': 150.0, 'category': 'Fruits'},
}
TAX_PERCENT = 5                          # tax added to every bill
DISCOUNT_RULES = [(500, 10), (300, 5)]   # (minimum bill, discount %), biggest first

categories = set()                       # a set keeps each category only once
for info in PRODUCTS.values():
    categories.add(info['category'])

print('Products loaded:', len(PRODUCTS))
print('Categories:', sorted(categories))
# Expected output:
# Products loaded: 5
# Categories: ['Bakery', 'Dairy', 'Fruits', 'Grains']
````

### Step 2: Show the Products

A function with one parameter and a `for` loop over the dictionary.

````python
# Step 2: Display all products
def show_products(products):
    """Print every product with its code and price."""
    print('Available products:')
    for code, info in products.items():
        name = info['name']
        price = info['price']
        print(f'  {code}: {name} - Rs. {price:.2f}')

show_products(PRODUCTS)
# Expected output:
# Available products:
#   P1: Rice (1 kg) - Rs. 60.00
#   P2: Milk (1 L) - Rs. 45.00
#   P3: Bread - Rs. 40.00
#   P4: Eggs (12) - Rs. 72.00
#   P5: Apples (1 kg) - Rs. 150.00
````

### Step 3: Add Items to the Cart

The cart is a dictionary: `{product code: quantity}`.
- `add_to_cart()` has a **default argument** (`qty=1`), checks its input with **`if`**, and **returns** `True` or `False`.
- `add_many()` uses **`*args`** to accept any number of product codes.
- Both functions change the cart dictionary that is passed in (see Part 4).

````python
# Step 3: Adding items, with checks
def add_to_cart(cart, code, qty=1):
    """Add qty of a product to the cart. Return True if it worked."""
    if code not in PRODUCTS:
        print(f'  Sorry, {code} is not a valid product code')
        return False
    if qty <= 0:
        print('  Quantity must be at least 1')
        return False
    cart[code] = cart.get(code, 0) + qty     # add to any quantity already in the cart
    name = PRODUCTS[code]['name']
    print(f'  Added {qty} x {name}')
    return True

def add_many(cart, *codes):
    """Add one of each product code given (any number of codes)."""
    for code in codes:
        add_to_cart(cart, code)

cart = {}
add_to_cart(cart, 'P1', 2)
add_to_cart(cart, 'P5')
add_many(cart, 'P2', 'P2', 'P3')
add_to_cart(cart, 'P9')
add_to_cart(cart, 'P4', 0)
print('Cart:', cart)
# Expected output:
#   Added 2 x Rice (1 kg)
#   Added 1 x Apples (1 kg)
#   Added 1 x Milk (1 L)
#   Added 1 x Milk (1 L)
#   Added 1 x Bread
#   Sorry, P9 is not a valid product code
#   Quantity must be at least 1
# Cart: {'P1': 2, 'P5': 1, 'P2': 2, 'P3': 1}
````

### Step 4: A Generator for the Bill Lines

`line_items()` **yields** one `(name, qty, amount)` tuple at a time, instead of building a whole list first.

````python
# Step 4: Generator that produces one bill line at a time
def line_items(cart):
    """Yield (name, qty, amount) for each item in the cart."""
    for code, qty in cart.items():
        info = PRODUCTS[code]
        yield info['name'], qty, info['price'] * qty

for name, qty, amount in line_items(cart):
    print(f'{name} x {qty} = Rs. {amount:.2f}')
# Expected output:
# Rice (1 kg) x 2 = Rs. 120.00
# Apples (1 kg) x 1 = Rs. 150.00
# Milk (1 L) x 2 = Rs. 90.00
# Bread x 1 = Rs. 40.00
````

### Step 5: Calculate the Bill

- `discount_percent()` loops over the rules and uses an **early `return`**.
- `calculate_bill()` has a **default argument** for the tax and **returns four values** at once.

````python
# Step 5: Discount, tax and total
def discount_percent(amount):
    """Return the discount % for a bill amount, using DISCOUNT_RULES."""
    for minimum, percent in DISCOUNT_RULES:
        if amount >= minimum:
            return percent
    return 0

def calculate_bill(cart, tax_percent=TAX_PERCENT):
    """Return subtotal, discount, tax and total for the cart."""
    subtotal = 0
    for name, qty, amount in line_items(cart):
        subtotal += amount
    discount = subtotal * discount_percent(subtotal) / 100
    tax = (subtotal - discount) * tax_percent / 100
    total = subtotal - discount + tax
    return subtotal, discount, tax, total

print(discount_percent(650), discount_percent(400), discount_percent(120))
subtotal, discount, tax, total = calculate_bill(cart)
print(subtotal, discount, tax, total)
# Expected output:
# 10 5 0
# 400.0 20.0 19.0 399.0
````

### Step 6: Print the Bill

- `print_row()` is a tiny helper, so the total lines are formatted in one place only.
- `print_bill()` takes the customer details as **`**kwargs`**, sorts the items with a **`lambda`**, and collects the categories in a **set**.

````python
# Step 6: Print a neat bill
def print_row(label, value):
    """Print one line of the totals, neatly aligned."""
    print(f'{label:<15} Rs. {value:>8.2f}')

def print_bill(cart, **customer):
    """Print the full bill. Customer details are given as keyword arguments."""
    if not cart:
        print('Cart is empty, nothing to bill')
        return
    print('====== SMART GROCERY BILL ======')
    for key, value in customer.items():
        print(f'{key.title()}: {value}')
    print('-' * 32)

    # Most expensive first: sort the generator's tuples by amount (index 2)
    items = sorted(line_items(cart), key=lambda item: item[2], reverse=True)
    for name, qty, amount in items:
        print(f'{name} x {qty} = Rs. {amount:.2f}')

    categories = set()
    for code in cart:
        categories.add(PRODUCTS[code]['category'])
    print('Categories:', ', '.join(sorted(categories)))   # join() glues the names with ', '
    print('-' * 32)

    subtotal, discount, tax, total = calculate_bill(cart)
    print_row('Subtotal', subtotal)
    if discount > 0:
        print_row(f'Discount ({discount_percent(subtotal)}%)', -discount)
    print_row(f'Tax ({TAX_PERCENT}%)', tax)
    print_row('TOTAL', total)
    print('=' * 32)

print_bill({})      # quick test with an empty cart
# Expected output: Cart is empty, nothing to bill
````

### Step 8: The Complete Interactive App

The menu runs in a `while True` loop. It reads choices with `input()`, converts the quantity with `int()`, chooses actions with `if-elif-else`, uses `continue` to skip bad input and `break` to finish.

The `cart` inside `run_shop()` is a **local** variable, so it is separate from the global `cart` used in Steps 3 and 7 (Part 8: scope).

````python
# Step 8: The full billing app (uses input - type your answers in the box)
def run_shop():
    """Run the shop menu until the customer checks out."""
    cart = {}                                               # local cart for this customer
    customer = input('Customer name: ').strip().title()    # strip() removes extra spaces, title() capitalises
    while True:
        print()
        print('1. Show products  2. Add item  3. View bill  4. Checkout')
        choice = input('Choose an option: ')
        if choice == '1':
            show_products(PRODUCTS)
        elif choice == '2':
            code = input('Product code: ').strip().upper()  # 'p5' becomes 'P5'
            qty_text = input('Quantity: ')
            if not qty_text.isdigit():                      # isdigit() is True only for whole numbers
                print('  Please enter a whole number')
                continue
            add_to_cart(cart, code, int(qty_text))
        elif choice == '3':
            print_bill(cart, name=customer)
        elif choice == '4':
            print_bill(cart, name=customer, payment='Cash')
            print('Thank you for shopping with us!')
            break
        else:
            print('  Invalid option, choose 1 to 4')

run_shop()
# Example run:
# Customer name: asha verma
#
# 1. Show products  2. Add item  3. View bill  4. Checkout
# Choose an option: 2
# Product code: p5
# Quantity: 2
#   Added 2 x Apples (1 kg)
#
# 1. Show products  2. Add item  3. View bill  4. Checkout
# Choose an option: 4
# ====== SMART GROCERY BILL ======
# Name: Asha Verma
# Payment: Cash
# --------------------------------
# Apples (1 kg) x 2 = Rs. 300.00
# Categories: Fruits
# --------------------------------
# Subtotal        Rs.   300.00
# Discount (5%)   Rs.   -15.00
# Tax (5%)        Rs.    14.25
# TOTAL           Rs.   299.25
# ================================
# Thank you for shopping with us!
````
