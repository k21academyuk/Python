# 1. Conditional Statements — `if` / `elif` / `else`

## What is a Conditional Statement?

A **conditional statement** lets your program **make a decision**. It evaluates a condition (an expression that is either `True` or `False`) and runs a block of code **only if** that condition is satisfied. This is exactly how humans make decisions every day — *"IF it is raining, THEN take an umbrella, ELSE wear sunglasses."*

### Key Characteristics

| Property | Value |
|---|---|
| Evaluates | A **boolean condition** (`True` / `False`) |
| Indentation | Python uses **indentation** (not `{ }`) to mark a block |
| `elif` | Short for "else if" — checks another condition if the first was `False` |
| `else` | The catch-all branch — runs when **no** condition above was `True` |
| Number of branches | Exactly **one** branch executes (or none, if there's no matching `elif`/`else`) |
| Nesting | You can place an `if` **inside** another `if` for multi-level decisions |

### 📐 Decision Flow — How Python Evaluates `if` / `elif` / `else`

diagram_2_if_elif_else.png



### Explanation of the Diagram

- Python checks conditions **top to bottom, in order**.
- The moment a condition is `True`, its block runs and **all remaining `elif`/`else` branches are skipped** — Python does not check them at all.
- If none of the `if`/`elif` conditions are `True`, the `else` block runs (if present). `else` is optional.
- After the matching block finishes, execution continues with whatever code comes **after** the entire `if` structure.

---

### 1.1 The Simple `if` Statement — Real-World Example: Voting Eligibility

The simplest form of conditional runs a block **only if** a condition is `True`. If it's `False`, Python simply skips the block — there is no "else" here.

````python
# ---- Real-World Example: Voting Eligibility Checker ----

age = 20
country = "India"

print(f"Checking voter eligibility for a {age}-year-old citizen of {country}...")

if age >= 18:
    print("✅ Eligible to vote!")

print("Eligibility check complete.")
````

**📋 Expected Output:**

```
Checking voter eligibility for a 20-year-old citizen of India...
✅ Eligible to vote!
Eligibility check complete.
```

### Code Explanation

- `age >= 18` is the **condition** — it evaluates to `True` or `False`.
- The indented `print()` line is the **if-block**; it only runs when the condition is `True`.
- Notice `"Eligibility check complete."` runs **regardless** — it's outside the `if` block, so it's not conditional at all.
- Try changing `age = 20` to `age = 15` and re-running: the ✅ line will simply not appear.

---

### 1.2 `if...else` — Real-World Example: Login Access Control

`if...else` gives you **exactly two** paths — one runs when the condition is `True`, the other when it's `False`. Exactly one of the two always executes.

````python
# ---- Real-World Example: Simple Login System ----

correct_password = "Py#2026Secure"
entered_password = "Py#2026Secure"

if entered_password == correct_password:
    print("🔓 Access granted. Welcome back!")
else:
    print("🔒 Access denied. Incorrect password.")

# ---- A second real-world use: Even / Odd Classifier for order IDs ----
order_id = 4587

if order_id % 2 == 0:
    print(f"Order #{order_id} routed to Warehouse A (even IDs)")
else:
    print(f"Order #{order_id} routed to Warehouse B (odd IDs)")
````

**📋 Expected Output:**

```
🔓 Access granted. Welcome back!
Order #4587 routed to Warehouse B (odd IDs)
```

### Code Explanation

- `==` **compares** two values for equality; it's different from `=`, which **assigns** a value.
- `order_id % 2 == 0` uses the **modulo operator `%`**, which returns the remainder of a division — a classic way to test even/odd.
- Because `if...else` is mutually exclusive, exactly one message prints in each block — never both, never neither.

---

### 1.3 `if...elif...else` Ladder — Real-World Example: Student Grading System

When there are **more than two** possible outcomes, chain conditions together with `elif` ("else if"). Python walks down the ladder and stops at the first `True` condition.

````python
# ---- Real-World Example: Exam Grading System ----

def get_grade(marks):
    if marks >= 90:
        return "A+ (Outstanding)"
    elif marks >= 80:
        return "A (Excellent)"
    elif marks >= 70:
        return "B (Very Good)"
    elif marks >= 60:
        return "C (Good)"
    elif marks >= 50:
        return "D (Pass)"
    else:
        return "F (Fail)"

students = {
    "Aarav": 95,
    "Diya": 82,
    "Kabir": 67,
    "Meera": 48,
}

for name, marks in students.items():
    grade = get_grade(marks)
    print(f"{name:8s} scored {marks:3d} -> Grade: {grade}")
````

**📋 Expected Output:**

```
Aarav    scored  95 -> Grade: A+ (Outstanding)
Diya     scored  82 -> Grade: A (Excellent)
Kabir    scored  67 -> Grade: C (Good)
Meera    scored  48 -> Grade: F (Fail)
```

### Code Explanation

- The conditions are ordered from **highest to lowest**. Since Python stops at the first `True` match, `marks >= 90` must be checked **before** `marks >= 80` — otherwise a 95 would incorrectly satisfy `marks >= 80` too early... but since we check the highest bar first, that's avoided.
- `elif` avoids writing multiple **independent** `if` statements (which would all be checked even after a match is found — wasteful and sometimes buggy).
- The `else` at the bottom is the **safety net**: it catches every case not covered above (here, marks below 50).
- We looped over a dictionary with `.items()` to grade every student in one pass — this previews the `for` loop covered in Section 3.

### 1.4 Nested `if` Statements — Real-World Example: Bank Loan Eligibility

A **nested if** is an `if` statement placed *inside* another `if`/`elif`/`else` block. It's used when a decision only makes sense **after** an outer condition is already satisfied — for example, checking income only for applicants who already meet the minimum age.

````python
# ---- Real-World Example: Loan Eligibility Checker ----

def check_loan_eligibility(age, monthly_income, credit_score):
    if age >= 21:
        print("Step 1: Age requirement met (21+).")
        if monthly_income >= 30000:
            print("Step 2: Income requirement met (>= 30,000).")
            if credit_score >= 700:
                print("Step 3: Credit score requirement met (>= 700).")
                print("✅ RESULT: Loan APPROVED.")
            else:
                print("Step 3: Credit score too low (< 700).")
                print("❌ RESULT: Loan REJECTED — improve credit score.")
        else:
            print("Step 2: Income too low (< 30,000).")
            print("❌ RESULT: Loan REJECTED — income requirement not met.")
    else:
        print("Step 1: Applicant is under 21.")
        print("❌ RESULT: Loan REJECTED — must be 21 or older.")

print("--- Applicant 1 ---")
check_loan_eligibility(age=28, monthly_income=45000, credit_score=750)

print("\n--- Applicant 2 ---")
check_loan_eligibility(age=25, monthly_income=20000, credit_score=680)
````

**📋 Expected Output:**

```
--- Applicant 1 ---
Step 1: Age requirement met (21+).
Step 2: Income requirement met (>= 30,000).
Step 3: Credit score requirement met (>= 700).
✅ RESULT: Loan APPROVED.

--- Applicant 2 ---
Step 1: Age requirement met (21+).
Step 2: Income too low (< 30,000).
❌ RESULT: Loan REJECTED — income requirement not met.
```

### Code Explanation

- Each inner `if` is only **reached** if the outer condition was `True` — the income check never even runs for a 19-year-old, because the outer `else` catches that case first.
- Notice the **indentation depth** grows with each nested level; Python uses this indentation to know which `if` a given `else` belongs to.
- Nesting is powerful but can get hard to read past 3–4 levels. A common real-world alternative is combining conditions with `and` — for example `if age >= 21 and monthly_income >= 30000 and credit_score >= 700:` — which is flatter and often clearer when you don't need step-by-step messages in between.

### 1.5 Real-World Capstone — ATM Cash Withdrawal Validation

Real programs usually combine **multiple conditions** with logical operators `and`, `or`, `not`. Here's how an ATM validates a withdrawal request before dispensing cash.

````python
# ---- Real-World Example: ATM Withdrawal Validation ----

account_balance = 15000
daily_limit = 25000
withdrawn_today = 20000

def validate_withdrawal(amount):
    print(f"Requested withdrawal: ₹{amount}")

    if amount <= 0:
        print("❌ Enter an amount greater than zero.")
    elif amount % 100 != 0:
        print("❌ Amount must be in multiples of ₹100.")
    elif amount > account_balance:
        print(f"❌ Insufficient balance. Available: ₹{account_balance}")
    elif (withdrawn_today + amount) > daily_limit:
        remaining = daily_limit - withdrawn_today
        print(f"❌ Daily limit exceeded. You can withdraw up to ₹{remaining} more today.")
    else:
        print(f"✅ Dispensing ₹{amount}. Please collect your cash.")

validate_withdrawal(5000)
print()
validate_withdrawal(9000)
print()
validate_withdrawal(250)
````

**📋 Expected Output:**

```
Requested withdrawal: ₹5000
✅ Dispensing ₹5000. Please collect your cash.

Requested withdrawal: ₹9000
❌ Daily limit exceeded. You can withdraw up to ₹5000 more today.

Requested withdrawal: ₹250
❌ Amount must be in multiples of ₹100.
```

### Code Explanation

| Check | What It Catches |
|---|---|
| `amount <= 0` | Zero or negative withdrawal requests |
| `amount % 100 != 0` | ATMs typically only dispense multiples of ₹100 |
| `amount > account_balance` | Not enough money in the account |
| `(withdrawn_today + amount) > daily_limit` | Would exceed the bank's daily withdrawal cap |
| `else` | Every check passed — safe to dispense cash |

- Each `elif` handles **one specific failure reason**, so the user gets a precise error message instead of a generic "denied".
- This is the `if/elif/else` **ladder pattern** applied to real validation logic — one of the most common uses of conditionals in production code (form validation, payment processing, access control, etc.).

### 1.6 Shorthand `if` — The Ternary (Conditional) Expression

When the decision is simple — just picking between two **values** — Python offers a compact one-line form: `value_if_true if condition else value_if_false`.

````python
# ---- Ternary / Conditional Expression ----

temperature = 38

# Long form
if temperature > 35:
    weather_advice = "Stay hydrated, it's hot!"
else:
    weather_advice = "Pleasant weather today."
print("Long form :", weather_advice)

# Short form (ternary) -- does exactly the same thing in one line
weather_advice_short = "Stay hydrated, it's hot!" if temperature > 35 else "Pleasant weather today."
print("Short form:", weather_advice_short)

# Real-world use: labeling stock status in an inventory report
stock_counts = {"USB-C Cable": 0, "Wireless Mouse": 14, "Laptop Stand": 3}
for item, qty in stock_counts.items():
    status = "OUT OF STOCK" if qty == 0 else ("LOW STOCK" if qty < 5 else "IN STOCK")
    print(f"{item:16s}: {status}")
````

**📋 Expected Output:**

```
Long form : Stay hydrated, it's hot!
Short form: Stay hydrated, it's hot!
USB-C Cable     : OUT OF STOCK
Wireless Mouse  : IN STOCK
Laptop Stand    : LOW STOCK
```

### Code Explanation

- The ternary expression `A if condition else B` reads almost like English: *"A if condition is true, otherwise B."*
- It's best used for **short, simple** value choices — assigning a label, a default, or a formatted string. For multi-line logic, a regular `if/else` is more readable.
- You can even **chain** ternaries (as in the stock example), but nesting more than one level usually hurts readability — prefer a full `if/elif/else` at that point.

# 2. The `while` Loop — Repeat *While* a Condition Holds

## What is a `while` Loop?

A **`while` loop** repeats a block of code **as long as** its condition remains `True`. It's the right tool when you **don't know in advance** how many times you need to repeat — you only know the *stopping condition* (e.g. "keep asking for a PIN until it's correct, or the user runs out of attempts").

### Key Characteristics

| Property | Value |
|---|---|
| Repeats while | Condition is `True` |
| Stops when | Condition becomes `False` (or a `break` is hit) |
| Best suited for | Unknown number of iterations, waiting for an event, retry logic |
| Risk | **Infinite loop** if the condition never becomes `False` |
| `else` clause | Optional — runs only if the loop finished **without** hitting `break` |

### 📐 Loop Flow — How a `while` Loop Executes

diagram_3_while_loop.png



### Explanation of the Diagram

- The condition is checked **before** each run of the loop body — if it's `False` on the very first check, the body never runs at all (zero iterations).
- Something **inside** the loop body must eventually change the condition's outcome (e.g. incrementing a counter, updating a flag) — otherwise you get an **infinite loop**.
- `break` (covered in Section 4) can exit a `while` loop early, even while the condition is still `True`.

---

### 2.1 Basic `while` Loop — Real-World Example: Rocket Launch Countdown

````python
# ---- Real-World Example: Countdown Timer ----

countdown = 5

print("Initiating launch sequence...")
while countdown > 0:
    print(f"T-minus {countdown}...")
    countdown -= 1   # same as: countdown = countdown - 1

print("🚀 Liftoff!")
````

**📋 Expected Output:**

```
Initiating launch sequence...
T-minus 5...
T-minus 4...
T-minus 3...
T-minus 2...
T-minus 1...
🚀 Liftoff!
```

### Code Explanation

- `countdown -= 1` is the **update step** — without it, `countdown > 0` would stay `True` forever and the program would hang in an infinite loop.
- The loop checks `countdown > 0` **before** every single print — the moment `countdown` becomes `0`, the loop stops and `"🚀 Liftoff!"` prints once, outside the loop.

---

### 2.2 Real-World Example: ATM PIN Retry System (Limited Attempts)

A very common real-world pattern: **keep retrying until success, but give up after a fixed number of attempts.** This needs *two* stopping conditions combined with `and`.

````python
# ---- Real-World Example: ATM PIN Verification with 3 Attempts ----

correct_pin = "4821"

# Simulating what a user might type in, attempt by attempt
# (in a real ATM this would come from the keypad via input())
simulated_attempts = ["1234", "9999", "4821"]

attempt_number = 0
max_attempts = 3
authenticated = False

while attempt_number < max_attempts and not authenticated:
    entered_pin = simulated_attempts[attempt_number]
    attempt_number += 1
    print(f"Attempt {attempt_number}: PIN entered = {entered_pin}")

    if entered_pin == correct_pin:
        authenticated = True
        print("✅ PIN correct. Access granted.")
    else:
        remaining = max_attempts - attempt_number
        print(f"❌ Incorrect PIN. Attempts remaining: {remaining}")

if not authenticated:
    print("\n🔒 Card blocked. Too many incorrect attempts.")
````

**📋 Expected Output:**

```
Attempt 1: PIN entered = 1234
❌ Incorrect PIN. Attempts remaining: 2
Attempt 2: PIN entered = 9999
❌ Incorrect PIN. Attempts remaining: 1
Attempt 3: PIN entered = 4821
✅ PIN correct. Access granted.
```

### Code Explanation

- The loop condition `attempt_number < max_attempts and not authenticated` combines **two** stopping reasons: running out of attempts, **or** succeeding early.
- `authenticated = True` inside the loop changes the condition's outcome — the next check of `not authenticated` becomes `False`, so the loop exits immediately on success without wasting the remaining attempts.
- The final `if not authenticated:` runs **after** the loop to handle the "ran out of attempts" case — this pattern (loop, then check *why* it ended) is extremely common in retry logic.

---

### 2.3 Real-World Example: Shopping Cart Accumulator (Sentinel-Controlled Loop)

Another classic `while` pattern: keep accumulating values **until a special "stop" signal** (called a *sentinel value*) appears — like a cashier scanning items until the customer says "that's all."

````python
# ---- Real-World Example: Grocery Checkout Running Total ----

# Simulating a scanner feed: item name, price -- "DONE" is the sentinel that ends checkout
scanned_items = [
    ("Milk", 60), ("Bread", 45), ("Eggs (dozen)", 90),
    ("Butter", 55), ("DONE", 0)
]

index = 0
total = 0
item_count = 0

print("🛒 Starting checkout...\n")
while scanned_items[index][0] != "DONE":
    name, price = scanned_items[index]
    total += price
    item_count += 1
    print(f"Scanned: {name:15s} ₹{price}")
    index += 1

print(f"\nItems scanned : {item_count}")
print(f"Total bill    : ₹{total}")
````

**📋 Expected Output:**

```
🛒 Starting checkout...

Scanned: Milk            ₹60
Scanned: Bread            ₹45
Scanned: Eggs (dozen)     ₹90
Scanned: Butter           ₹55

Items scanned : 4
Total bill    : ₹250
```

### Code Explanation

- `"DONE"` is a **sentinel value** — a special marker that signals "stop", rather than being real data to process.
- The condition `scanned_items[index][0] != "DONE"` is checked **before** each scan — as soon as `"DONE"` is reached, the loop exits without processing it.
- This pattern mirrors real checkout systems, chatbots reading messages until `"quit"`, or file readers processing lines until an end marker.

---

### 2.4 The `while...else` Clause

Python loops support an optional `else` block — a feature many languages don't have. It runs **only if the loop finished naturally** (condition became `False`), and is **skipped if the loop was exited via `break`**.

````python
# ---- Real-World Example: Searching a Print Queue for a Specific Job ----

print_queue = ["Report.pdf", "Invoice.docx", "Photo.png", "Resume.pdf"]
target_job = "Presentation.pptx"

index = 0
while index < len(print_queue):
    if print_queue[index] == target_job:
        print(f"✅ Found '{target_job}' at position {index}. Printing now...")
        break
    index += 1
else:
    # Runs only because the loop completed WITHOUT finding the job (no break)
    print(f"❌ '{target_job}' not found in the print queue. Nothing to print.")

print("Queue check finished.")
````

**📋 Expected Output:**

```
❌ 'Presentation.pptx' not found in the print queue. Nothing to print.
Queue check finished.
```

### Code Explanation

- The `else` here belongs to the **`while` loop**, not to an `if` — it's easy to confuse the two at first glance.
- It runs when the loop's condition becomes `False` **naturally** — i.e., we scanned the entire queue and never hit `break`.
- If `target_job` **had** been found, `break` would fire, and the `else` block would be **skipped entirely** — this is the whole point of `while...else`: it answers *"did the loop find what it was looking for, or not?"* without needing a separate flag variable.

---

### ⚠️ A Word of Caution — Infinite Loops

If the condition in a `while` loop **never** becomes `False`, the loop runs forever and your program freezes. This is one of the most common bugs beginners write. Always double-check that something inside the loop moves the condition toward becoming `False`.

```python
# DANGER — infinite loop, never run this without a break condition:
# count = 0
# while count < 5:
#     print(count)
#     # forgot to increment count!  count never reaches 5.
```

If you ever accidentally trigger an infinite loop in a real notebook, use the **Interrupt Kernel** (stop button) to halt execution.

# 3. The `for` Loop — Iterating Over a Known Sequence

## What is a `for` Loop?

A **`for` loop** repeats a block of code **once for each item** in a sequence (a list, string, range, dictionary, etc). Unlike `while`, you don't manage a counter or condition manually — Python handles that for you. Use `for` whenever you know **what you're iterating over**, even if you don't know exactly how many items there are.

### Key Characteristics

| Property | Value |
|---|---|
| Repeats over | Every item in an **iterable** (list, tuple, string, dict, range, set, ...) |
| Loop variable | Automatically takes each item's value, one at a time |
| Best suited for | Known collections, a fixed number of repetitions, transforming data |
| `range(n)` | Generates numbers `0, 1, ..., n-1` — used for "repeat n times" |
| `else` clause | Optional — runs only if the loop finished **without** hitting `break` |

### 📐 Loop Flow — How a `for` Loop Executes

image.png



### Explanation of the Diagram

- The loop variable (`fruit`) is **reassigned** at the start of each iteration to the next item in the sequence.
- There's no manual condition to maintain and no risk of an infinite loop — the loop automatically stops after the **last** item.
- The number of iterations is exactly the number of items in the sequence (here, 3).

---

### 3.1 Iterating Over Lists, Strings, and `range()` — Real-World Example: Order Processing

````python
# ---- Real-World Example: Processing a List of Orders ----

pending_orders = ["Order#101", "Order#102", "Order#103"]

print("Processing orders:")
for order in pending_orders:
    print(f"  -> Shipping {order}")

# Iterating over a string -- Real-World Example: Validating a product code character by character
product_code = "AB12"
digit_count = 0
for char in product_code:
    if char.isdigit():
        digit_count += 1
print(f"\nProduct code '{product_code}' contains {digit_count} digit(s).")

# Iterating with range() -- Real-World Example: Sending 3 reminder emails, day by day
print("\nScheduling reminder emails:")
for day in range(1, 4):
    print(f"  -> Reminder email #{day} scheduled for Day {day}")
````

**📋 Expected Output:**

```
Processing orders:
  -> Shipping Order#101
  -> Shipping Order#102
  -> Shipping Order#103

Product code 'AB12' contains 2 digit(s).

Scheduling reminder emails:
  -> Reminder email #1 scheduled for Day 1
  -> Reminder email #2 scheduled for Day 2
  -> Reminder email #3 scheduled for Day 3
```

### Code Explanation

- Looping over a **list** gives you each element directly — no indexing needed.
- Looping over a **string** gives you each character, one at a time — handy for validation, counting, or building new strings character-by-character.
- `range(1, 4)` generates `1, 2, 3` — the **stop** value (`4`) is never included. `range(n)` alone generates `0` to `n-1`, which is why it's perfect for "repeat exactly n times."

---

### 3.2 `for` with `range()` — Real-World Example: Multiplication Table Generator

A very common use of `for` + `range()` is generating a fixed, predictable sequence of calculations — like a multiplication table used in a school learning app.

````python
# ---- Real-World Example: Multiplication Table Generator ----

def print_table(number, up_to=10):
    print(f"Multiplication table for {number}:")
    for i in range(1, up_to + 1):
        print(f"  {number} x {i:2d} = {number * i}")

print_table(7, up_to=5)
````

**📋 Expected Output:**

```
Multiplication table for 7:
  7 x  1 = 7
  7 x  2 = 14
  7 x  3 = 21
  7 x  4 = 28
  7 x  5 = 35
```

### Code Explanation

- `range(1, up_to + 1)` runs from `1` through `up_to` **inclusive** — we add `1` to the stop value because `range()`'s upper bound is exclusive.
- `i:2d` inside the f-string pads the number to 2 characters wide, so the output aligns neatly — a small but common formatting touch in real reports.
- Wrapping this in a function (`print_table`) lets us reuse the same loop logic for **any** number — a hallmark of well-organized real-world code.

---

### 3.3 Nested `for` Loops — Real-World Example: Cinema Seat Map

A **nested loop** is a loop inside another loop. The inner loop completes **all** of its iterations for every single iteration of the outer loop — perfect for anything grid-shaped: seating charts, spreadsheets, image pixels, game boards.

````python
# ---- Real-World Example: Generating a Cinema Seating Chart ----

rows = ["A", "B", "C"]
seats_per_row = 5
booked_seats = {"A3", "B1", "C5"}

print("🎬 Cinema Seat Map (X = booked, O = available)\n")
for row in rows:
    row_display = ""
    for seat_num in range(1, seats_per_row + 1):
        seat_id = f"{row}{seat_num}"
        symbol = "X" if seat_id in booked_seats else "O"
        row_display += symbol + " "
    print(f"Row {row}: {row_display}")
````

**📋 Expected Output:**

```
🎬 Cinema Seat Map (X = booked, O = available)

Row A: O O X O O
Row B: X O O O O
Row C: O O O O X
```

### Code Explanation

- The **outer loop** (`for row in rows`) runs 3 times — once per row.
- The **inner loop** (`for seat_num in range(...)`) runs 5 times **for each** outer iteration — so the inner `print`-building logic actually executes `3 x 5 = 15` times in total.
- This "outer picks the row, inner fills the row" pattern is exactly how 2-D grids, tables, and matrices are built and printed in real applications.

---

### 3.4 `enumerate()` and `zip()` — Real-World Example: Processing Student Records

Two extremely common companions to `for`:
- **`enumerate()`** gives you both the **index** and the **value** while looping — useful for numbering things.
- **`zip()`** lets you loop over **two or more sequences in parallel** — useful for combining related lists.

````python
# ---- Real-World Example: Report Card Generator ----

student_names = ["Ishaan", "Priya", "Rohan"]
student_scores = [88, 76, 92]

print("Rank | Student | Score")
print("-" * 26)

# enumerate() -- get a running rank number automatically
# zip() -- walk through names and scores together, pair by pair
for rank, (name, score) in enumerate(zip(student_names, student_scores), start=1):
    print(f"{rank:>4} | {name:<7} | {score}")
````

**📋 Expected Output:**

```
Rank | Student | Score
--------------------------
   1 | Ishaan  | 88
   2 | Priya   | 76
   3 | Rohan   | 92
```

### Code Explanation

- `zip(student_names, student_scores)` pairs up items **positionally**: `("Ishaan", 88)`, `("Priya", 76)`, `("Rohan", 92)` — without it, you'd need manual indexing like `student_names[i]`.
- `enumerate(..., start=1)` adds a counter to each pair, starting at `1` instead of the default `0` — perfect for human-friendly rankings.
- This combination avoids error-prone manual index management (`for i in range(len(student_names))`) while staying just as powerful.

### 3.5 `for...else` — Real-World Example: Prime Number Checker

Just like `while`, a `for` loop can have an `else` block that runs **only if the loop completes without hitting `break`.** This is a natural fit for "search and report if nothing was found" logic.

````python
# ---- Real-World Example: Checking Whether a Number is Prime ----

def check_prime(n):
    if n < 2:
        print(f"{n} is not a prime number.")
        return

    for divisor in range(2, n):
        if n % divisor == 0:
            print(f"{n} is NOT prime (divisible by {divisor}).")
            break
    else:
        # Runs only if the loop never found a divisor (never hit break)
        print(f"{n} IS a prime number.")

for number in [17, 18, 23]:
    check_prime(number)
````

**📋 Expected Output:**

```
17 IS a prime number.
18 is NOT prime (divisible by 2).
23 IS a prime number.
```

### Code Explanation

- The inner `for` loop tries every possible divisor from `2` up to `n - 1`. The moment it finds one that divides evenly, the number **can't** be prime — it prints the reason and `break`s immediately.
- The `else` block only runs if the loop finished the **entire** range **without ever breaking** — meaning no divisor was found, so the number must be prime.
- This avoids a separate `is_prime = True` flag variable that you'd otherwise need to track and check after the loop — `for...else` does that bookkeeping for you.

# 4. Jumping Statements — `break`, `continue`, `pass`

## What are Jumping Statements?

**Jumping statements** alter the normal top-to-bottom flow of a loop (or, in the case of `pass`, any block) from the **inside**. They let you exit early, skip an iteration, or explicitly say "do nothing here."

### Key Characteristics

| Statement | What It Does | Typical Use |
|---|---|---|
| `break` | **Immediately exits** the nearest enclosing loop entirely | Stop searching once found; abort on error |
| `continue` | **Skips the rest** of the current iteration and jumps to the next one | Skip invalid/unwanted items but keep looping |
| `pass` | Does **absolutely nothing** — a placeholder | Empty function/class body; "TODO" stub; ignore a specific case on purpose |

### 📐 How `break` and `continue` Change Loop Flow
image.png



### Explanation of the Diagram

- `break` is **drastic** — once it fires, the loop is over, no matter how many items were left.
- `continue` is **gentle** — it skips just the current round and the loop keeps going with the next item.
- Both only affect the **nearest enclosing loop** — in nested loops, `break`/`continue` inside the inner loop does not affect the outer loop at all.

---

### 4.1 `break` — Real-World Example: Linear Search (Stop as Soon as You Find It)

`break` is the natural choice whenever you're **searching** for something — there's no point checking the rest once you've found it.

````python
# ---- Real-World Example: Searching a Customer Database by ID ----

customers = [
    {"id": "C101", "name": "Ananya Rao"},
    {"id": "C102", "name": "Vikram Singh"},
    {"id": "C103", "name": "Neha Kapoor"},
    {"id": "C104", "name": "Sameer Ali"},
]

search_id = "C103"
found_customer = None

for customer in customers:
    print(f"Checking {customer['id']}...")
    if customer["id"] == search_id:
        found_customer = customer
        break   # No need to check the remaining customers -- stop right here

if found_customer:
    print(f"\n✅ Found: {found_customer['name']} ({found_customer['id']})")
else:
    print("\n❌ Customer not found.")
````

**📋 Expected Output:**

```
Checking C101...
Checking C102...
Checking C103...

✅ Found: Neha Kapoor (C103)
```

### Code Explanation

- Notice `"Checking C104..."` **never prints** — `break` stopped the loop the instant a match was found at `C103`, skipping the remaining customer.
- Without `break`, the loop would keep checking every remaining customer even after finding the right one — wasted work that matters a lot on large real-world datasets (millions of records).
- This "search and stop" pattern is the single most common real-world use of `break`.

---

### 4.2 `continue` — Real-World Example: Data Validation (Skip Bad Entries, Keep Going)

`continue` is the natural choice when you want to **skip** certain items but still process everything else — like ignoring invalid entries in a batch of sensor readings.

````python
# ---- Real-World Example: Cleaning a Batch of Temperature Sensor Readings ----

raw_readings = [22.5, -999, 24.1, None, 23.8, -999, 21.9]  # -999 and None mean "faulty reading"

valid_readings = []

for reading in raw_readings:
    if reading is None or reading == -999:
        print(f"⚠️  Skipping faulty reading: {reading}")
        continue   # skip this one, move straight to the next reading

    valid_readings.append(reading)
    print(f"✅ Accepted reading: {reading}°C")

average = sum(valid_readings) / len(valid_readings)
print(f"\nValid readings used: {len(valid_readings)} / {len(raw_readings)}")
print(f"Average temperature: {average:.2f}°C")
````

**📋 Expected Output:**

```
✅ Accepted reading: 22.5°C
⚠️  Skipping faulty reading: -999
✅ Accepted reading: 24.1°C
⚠️  Skipping faulty reading: None
✅ Accepted reading: 23.8°C
⚠️  Skipping faulty reading: -999
✅ Accepted reading: 21.9°C

Valid readings used: 5 / 7
Average temperature: 23.06°C
```

### Code Explanation

- `continue` jumps straight back to the top of the loop for the **next** reading — the lines `valid_readings.append(...)` and its `print` never execute for faulty entries.
- Crucially, the loop **keeps going** after a skip — unlike `break`, which would have stopped processing entirely after the first `-999`.
- This is exactly how real data-cleaning pipelines work: silently (or noisily, as here) skip bad rows while still processing the good ones.

### 4.3 `pass` — Real-World Example: Placeholder for Work-in-Progress Code

`pass` does **nothing at all** — it's a no-op. Python needs it because a block (after `if`, `for`, `while`, `def`, `class`, etc.) can never be **empty**; `pass` satisfies that requirement while you decide what to actually write later.

````python
# ---- Real-World Example: Stubbing Out Functions During Development ----

def send_email_notification(user):
    pass  # TODO: implement email-sending logic next sprint

def send_sms_notification(user):
    pass  # TODO: integrate with SMS gateway

def process_notifications(users):
    for user in users:
        print(f"Processing notifications for {user}...")
        send_email_notification(user)   # safe to call even though it's not implemented yet
        send_sms_notification(user)
    print("All notification stubs called without errors.")

process_notifications(["Aditi", "Karan"])

print()

# Real-World Example: Deliberately ignoring one specific case in a loop
transactions = [120, -50, 0, 340, -10]

for amount in transactions:
    if amount == 0:
        pass   # Zero-amount transactions are valid but intentionally ignored for reporting
    elif amount > 0:
        print(f"Credit: +{amount}")
    else:
        print(f"Debit: {amount}")
````

**📋 Expected Output:**

```
Processing notifications for Aditi...
Processing notifications for Karan...
All notification stubs called without errors.

Credit: +120
Debit: -50
Credit: +340
Debit: -10
```

### Code Explanation

- Without `pass`, `def send_email_notification(user):` followed by nothing would raise an `IndentationError` — Python requires **something** inside every block.
- `pass` is a genuine, deliberate **"do nothing"** — different from `continue` (which still moves the loop forward) or `break` (which exits). In the transactions example, the `amount == 0` branch is intentionally a no-op; execution simply falls through to the next loop iteration afterward, same as normal.
- Common real-world uses: stubbing out functions/classes before they're implemented, deliberately ignoring one branch of an `if`, or writing an empty exception handler while prototyping (`except ValueError: pass`) — though production code usually replaces these with real logic before shipping.

# 5. Capstone Project — ATM Menu-Driven Simulation

Real programs rarely use just one control-flow tool in isolation — they **combine** them. This capstone brings together everything from this lab into a single menu-driven ATM simulator:

- A `while True` loop keeps the menu running until the user chooses to exit.
- `if / elif / else` decides which action to take based on the menu choice.
- `continue` skips straight back to the menu after showing an error, without crashing.
- `break` exits the ATM loop cleanly when the user chooses "Exit".
- A `for` loop displays the transaction history at the end.

Since this notebook can't take live keyboard input, we **simulate** a user by feeding in a pre-written list of menu choices, one at a time — exactly the values a real user might type at the keypad.

````python
# ---- Capstone: ATM Menu-Driven Simulation ----

balance = 10000
transaction_history = []

# Simulated user input, as if typed one at a time on the ATM keypad
simulated_inputs = iter([
    "1",        # Check balance
    "2", "5000",   # Deposit 5000
    "3", "50000",  # Try to withdraw more than balance -> should fail gracefully
    "3", "2000",   # Withdraw 2000 -> should succeed
    "9",        # Invalid menu option
    "4",        # View transaction history
    "5",        # Exit
])

def get_input(prompt):
    value = next(simulated_inputs)
    print(f"{prompt}{value}")   # echo the simulated input like a real keypad would show it
    return value

print("======  🏧  WELCOME TO PYTHON BANK ATM  ======\n")

while True:
    print("\n--- MENU ---")
    print("1. Check Balance   2. Deposit   3. Withdraw   4. Transaction History   5. Exit")
    choice = get_input("Enter choice: ")

    if choice == "1":
        print(f"💰 Current balance: ₹{balance}")

    elif choice == "2":
        amount = int(get_input("Enter deposit amount: "))
        balance += amount
        transaction_history.append(f"Deposited ₹{amount}")
        print(f"✅ Deposit successful. New balance: ₹{balance}")

    elif choice == "3":
        amount = int(get_input("Enter withdrawal amount: "))
        if amount > balance:
            print(f"❌ Insufficient funds. Available: ₹{balance}")
            continue   # skip straight back to the menu -- no transaction recorded
        balance -= amount
        transaction_history.append(f"Withdrew ₹{amount}")
        print(f"✅ Withdrawal successful. New balance: ₹{balance}")

    elif choice == "4":
        print("📜 Transaction History:")
        if not transaction_history:
            print("  (no transactions yet)")
        for i, txn in enumerate(transaction_history, start=1):
            print(f"  {i}. {txn}")

    elif choice == "5":
        print("👋 Thank you for banking with us. Goodbye!")
        break   # exit the while-True loop entirely

    else:
        print("❌ Invalid option. Please choose between 1 and 5.")
        continue

print(f"\nFinal balance: ₹{balance}")
````

# 6. Summary Cheat Sheet

| Construct | Purpose | Runs Body | Stops When |
|---|---|---|---|
| `if` | Run code only if a condition is `True` | 0 or 1 times | N/A |
| `if / else` | Choose between exactly two paths | 1 time (one branch) | N/A |
| `if / elif / else` | Choose between many paths | 1 time (one branch) | First `True` condition wins |
| `while` | Repeat while a condition holds | 0+ times | Condition becomes `False`, or `break` |
| `for` | Repeat once per item in a sequence | Exactly `len(sequence)` times | Sequence exhausted, or `break` |
| `break` | Exit the current loop immediately | — | Fires unconditionally when reached |
| `continue` | Skip to the next iteration | — | Fires unconditionally when reached |
| `pass` | Do nothing (syntactic placeholder) | — | Never "fires" — it's a no-op |

### Quick Decision Guide

- **Need to pick one path out of several?** → `if` / `elif` / `else`
- **Don't know how many times you'll repeat, only the stopping condition?** → `while`
- **Know exactly what you're iterating over (a list, string, range, etc.)?** → `for`
- **Want to stop a loop the moment something is found/fails?** → `break`
- **Want to skip just one problematic item but keep looping?** → `continue`
- **Need an empty block for now (stub, TODO, deliberate no-op)?** → `pass`
