# Part 1: Lists

A **list** is like a shopping list: an ordered collection of items that you can add to, remove from and change at any time.

- Written with **square brackets** `[ ]`, items separated by commas
- **Ordered:** items stay in the order you put them in
- **Changeable (mutable):** you can add, remove or update items
- **Allows duplicates:** the same value can appear more than once
- **Mixed types:** one list can hold numbers, text, True/False and even other lists

## 1.1 Creating a List

Put items inside square brackets, separated by commas. You can also start with an empty list and fill it later, or turn other things into a list with `list()`.

**Syntax:**
```python
my_list = [item1, item2, item3]
```

````python
# Example 1: A list of text (strings)
fruits = ['apple', 'banana', 'mango']
print(fruits)
# Expected output: ['apple', 'banana', 'mango']
````

````python
# Example 2: A list of numbers, and a list with mixed data types
marks = [85, 90, 78, 92]
student = ['Asha', 21, 8.5, True]
print(marks)
print(student)
# Expected output:
# [85, 90, 78, 92]
# ['Asha', 21, 8.5, True]
````

````python
# Example 3: An empty list, and a list made from a word using list()
empty = []
letters = list('hello')
print(empty)
print(letters)
# Expected output:
# []
# ['h', 'e', 'l', 'l', 'o']
````

## 1.2 Indexing: Getting One Item

Every item in a list has a position number called an **index**. Python starts counting from **0**, not 1.

```
fruits = ['apple', 'banana', 'mango', 'grapes']
Index:       0        1         2        3
Negative:   -4       -3        -2       -1
```

- A **positive index** counts from the start (0 is the first item)
- A **negative index** counts from the end (-1 is the last item)
- Asking for an index that does not exist gives an `IndexError`

**Syntax:** `my_list[index]`

````python
# Example 1: Positive index (counting from the start)
fruits = ['apple', 'banana', 'mango', 'grapes']
print(fruits[0])   # first item
print(fruits[2])   # third item
# Expected output:
# apple
# mango
````

````python
# Example 2: Negative index (counting from the end)
fruits = ['apple', 'banana', 'mango', 'grapes']
print(fruits[-1])   # last item
print(fruits[-2])   # second last item
# Expected output:
# grapes
# mango
````

````python
# Example 3: Using an index that does not exist
fruits = ['apple', 'banana', 'mango', 'grapes']
# Valid indexes are 0 to 3. Remove the # from the next line to see the error:
# print(fruits[10])
# Expected output (after removing #): IndexError: list index out of range
````

## 1.3 Slicing: Getting a Part of the List

Slicing cuts out a piece of a list, like cutting a slice from a loaf of bread.

**Syntax:**
```python
my_list[start:stop:step]
```
- **start**: where to begin (included). Leave it empty to start from the beginning.
- **stop**: where to end (**not included**). Leave it empty to go to the end.
- **step**: how many places to jump each time (default is 1).

Slicing always gives you a **new list**; the original list does not change.

````python
# Example 1: Items from index 1 up to (but not including) index 4
nums = [10, 20, 30, 40, 50, 60]
print(nums[1:4])
# Expected output: [20, 30, 40]
````

````python
# Example 2: Leaving out start or stop
nums = [10, 20, 30, 40, 50, 60]
print(nums[:3])    # first 3 items
print(nums[3:])    # from index 3 to the end
print(nums[-2:])   # last 2 items
# Expected output:
# [10, 20, 30]
# [40, 50, 60]
# [50, 60]
````

````python
# Example 3: Using step
nums = [10, 20, 30, 40, 50, 60]
print(nums[::2])    # every 2nd item
print(nums[::-1])   # a step of -1 reverses the list
# Expected output:
# [10, 30, 50]
# [60, 50, 40, 30, 20, 10]
````

## 1.4 Changing (Updating) Items

Lists are changeable, so you can replace an item by giving its index a new value. You can also replace several items at once using a slice.

**Syntax:** `my_list[index] = new_value`

````python
# Example 1: Change one item using its index
fruits = ['apple', 'banana', 'mango']
fruits[1] = 'orange'
print(fruits)
# Expected output: ['apple', 'orange', 'mango']
````

````python
# Example 2: Change the last item using a negative index
marks = [85, 90, 78]
marks[-1] = 80
print(marks)
# Expected output: [85, 90, 80]
````

````python
# Example 3: Change several items at once using a slice
nums = [1, 2, 3, 4, 5]
nums[1:3] = [20, 30]
print(nums)
# Expected output: [1, 20, 30, 4, 5]
````

## 1.5 append(): Add One Item at the End

`append()` adds **one** item to the **end** of the list, like a person joining the back of a queue.

**Syntax:** `my_list.append(item)`

````python
# Example 1: Add a fruit to the end
fruits = ['apple', 'banana']
fruits.append('mango')
print(fruits)
# Expected output: ['apple', 'banana', 'mango']
````

````python
# Example 2: Build a list step by step, starting from an empty list
scores = []
scores.append(50)
scores.append(75)
print(scores)
# Expected output: [50, 75]
````

````python
# Example 3: Appending a list adds it as ONE item (a list inside a list)
nums = [1, 2]
nums.append([3, 4])
print(nums)
print(len(nums))   # len() counts the items
# Expected output:
# [1, 2, [3, 4]]
# 3
````

## 1.6 extend(): Add Many Items at the End

`extend()` takes each item from another collection and adds them **one by one** to the end of the list.

**append vs extend:** `append([3, 4])` adds the whole list as a single item, while `extend([3, 4])` adds 3 and 4 as two separate items.

**Syntax:** `my_list.extend(another_collection)`

````python
# Example 1: Add several fruits at once
fruits = ['apple', 'banana']
fruits.extend(['mango', 'grapes'])
print(fruits)
# Expected output: ['apple', 'banana', 'mango', 'grapes']
````

````python
# Example 2: Compare append and extend side by side
a = [1, 2]
b = [1, 2]
a.append([3, 4])
b.extend([3, 4])
print(a)
print(b)
# Expected output:
# [1, 2, [3, 4]]
# [1, 2, 3, 4]
````

````python
# Example 3: extend() works with any collection, even a string
letters = ['a', 'b']
letters.extend('cd')
print(letters)
# Expected output: ['a', 'b', 'c', 'd']
````

## 1.7 insert(): Add an Item at a Specific Position

`insert()` puts an item exactly where you want it. The items after that position shift one place to the right.

**Syntax:** `my_list.insert(index, item)`

````python
# Example 1: Insert at index 1 (second position)
fruits = ['apple', 'mango']
fruits.insert(1, 'banana')
print(fruits)
# Expected output: ['apple', 'banana', 'mango']
````

````python
# Example 2: Insert at the very beginning (index 0)
queue = ['Ravi', 'Meena']
queue.insert(0, 'Asha')
print(queue)
# Expected output: ['Asha', 'Ravi', 'Meena']
````

````python
# Example 3: If the index is bigger than the list, the item goes to the end
nums = [1, 2, 3]
nums.insert(100, 4)
print(nums)
# Expected output: [1, 2, 3, 4]
````

## 1.8 pop(): Remove an Item by Position (and Get It Back)

`pop()` removes an item **and hands it back to you**, so you can store it in a variable.
- `pop()` with nothing inside the brackets removes the **last** item
- `pop(index)` removes the item at that index

**Syntax:** `removed_item = my_list.pop(index)`

````python
# Example 1: Remove the last item
fruits = ['apple', 'banana', 'mango']
removed = fruits.pop()
print(removed)
print(fruits)
# Expected output:
# mango
# ['apple', 'banana']
````

````python
# Example 2: Remove the item at index 0
fruits = ['apple', 'banana', 'mango']
first = fruits.pop(0)
print(first)
print(fruits)
# Expected output:
# apple
# ['banana', 'mango']
````

## 1.9 remove(): Remove an Item by Value

`remove()` deletes an item when you know **what** it is but not **where** it is.
- Only the **first** match is removed
- If the item is not in the list, you get a `ValueError`

**pop vs remove:** `pop()` uses the position (index); `remove()` uses the value.

**Syntax:** `my_list.remove(value)`

````python
# Example 1: Remove a fruit by its name
fruits = ['apple', 'banana', 'mango']
fruits.remove('banana')
print(fruits)
# Expected output: ['apple', 'mango']
````

````python
# Example 2: Only the FIRST match is removed
nums = [5, 3, 5, 7]
nums.remove(5)
print(nums)
# Expected output: [3, 5, 7]
````

````python
# Example 3: Check with 'in' first, to avoid an error if the item is missing
fruits = ['apple', 'mango']
if 'kiwi' in fruits:
    fruits.remove('kiwi')
else:
    print('kiwi is not in the list')
# Expected output: kiwi is not in the list
````

## 1.10 del: Delete by Index or Slice

`del` is a Python keyword (not a list method). It deletes an item or a whole slice by position. Unlike `pop()`, it does not give the item back.

**Syntax:** `del my_list[index]` or `del my_list[start:stop]`

````python
# Example 1: Delete one item by index
fruits = ['apple', 'banana', 'mango']
del fruits[1]
print(fruits)
# Expected output: ['apple', 'mango']
````

````python
# Example 2: Delete a slice (several items at once)
nums = [10, 20, 30, 40, 50]
del nums[1:3]
print(nums)
# Expected output: [10, 40, 50]
````

## 1.11 clear(): Empty the List

`clear()` removes **all** items. The list still exists, but it is now empty, like emptying a box without throwing the box away.

**Syntax:** `my_list.clear()`

````python
# Example 1: Empty a shopping cart
cart = ['pen', 'notebook', 'bag']
cart.clear()
print(cart)
# Expected output: []
````

````python
# Example 2: After clear(), you can keep using the same list
cart = ['pen', 'notebook']
cart.clear()
cart.append('pencil')
print(cart)
# Expected output: ['pencil']
````

## 1.12 index() and count(): Find and Count Items

- `index(item)` tells you the **position** of the first match (error if the item is not there)
- `count(item)` tells you **how many times** an item appears (0 if it is not there)

**Syntax:** `my_list.index(item)` and `my_list.count(item)`

````python
# Example 1: Find the position of an item
fruits = ['apple', 'banana', 'mango']
print(fruits.index('mango'))
# Expected output: 2
````

````python
# Example 2: Count how many times a value appears
votes = ['yes', 'no', 'yes', 'yes', 'no']
print(votes.count('yes'))
print(votes.count('maybe'))   # not in the list, so 0
# Expected output:
# 3
# 0
````

````python
# Example 3: index() returns only the FIRST match
nums = [4, 7, 4, 9]
print(nums.index(4))
# Expected output: 0
````

## 1.13 sort() and sorted(): Arrange Items in Order

- `my_list.sort()` sorts the list **itself** (the original list changes)
- `sorted(my_list)` gives you a **new sorted list** and leaves the original untouched
- Add `reverse=True` to either one for descending order (biggest first)
- Text is sorted alphabetically

**Common mistake:** `sort()` returns nothing, so `marks = marks.sort()` turns `marks` into `None`. Just write `marks.sort()`.

````python
# Example 1: Sort numbers from smallest to largest
marks = [78, 92, 85, 60]
marks.sort()
print(marks)
# Expected output: [60, 78, 85, 92]
````

````python
# Example 2: Descending order, and sorting text alphabetically
marks = [78, 92, 85, 60]
marks.sort(reverse=True)
print(marks)

names = ['Ravi', 'Asha', 'Meena']
names.sort()
print(names)
# Expected output:
# [92, 85, 78, 60]
# ['Asha', 'Meena', 'Ravi']
````

````python
# Example 3: sorted() keeps the original list unchanged
marks = [78, 92, 85]
new_marks = sorted(marks)
print(new_marks)
print(marks)
# Expected output:
# [78, 85, 92]
# [78, 92, 85]
````

## 1.14 reverse(): Flip the Order

`reverse()` flips the list back to front. It does **not** sort; it only turns the current order around.

**Syntax:** `my_list.reverse()`

````python
# Example 1: Reverse a list
nums = [1, 2, 3, 4]
nums.reverse()
print(nums)
# Expected output: [4, 3, 2, 1]
````

````python
# Example 2: reverse() flips the order, it does not sort
nums = [3, 1, 2]
nums.reverse()
print(nums)
# Expected output: [2, 1, 3]
````

## 1.15 len(), min(), max() and sum(): Quick Facts About a List

These built-in functions look at the whole list:
- `len()` counts the items
- `min()` and `max()` find the smallest and largest
- `sum()` adds up all the numbers (works with numbers only)

````python
# Example 1: Facts about a list of marks
marks = [78, 92, 85, 60]
print(len(marks))
print(min(marks))
print(max(marks))
print(sum(marks))
# Expected output:
# 4
# 60
# 92
# 315
````

````python
# Example 2: Calculate the average
marks = [78, 92, 85, 60]
average = sum(marks) / len(marks)
print(average)
# Expected output: 78.75
````

## 1.16 in and not in: Check if an Item Exists

`in` answers a yes/no question: *is this item in the list?* It gives back `True` or `False`. `not in` asks the opposite.

**Syntax:** `item in my_list`

````python
# Example 1: Check membership
fruits = ['apple', 'banana', 'mango']
print('apple' in fruits)
print('kiwi' in fruits)
print('kiwi' not in fruits)
# Expected output:
# True
# False
# True
````

````python
# Example 2: Use 'in' inside an if statement
allowed_users = ['asha', 'ravi']
user = 'ravi'
if user in allowed_users:
    print('Access granted')
# Expected output: Access granted
````

## 1.17 Joining (+) and Repeating (*) Lists

- `+` joins two lists into one new list
- `*` repeats a list a number of times

````python
# Example 1: Join two lists
list1 = [1, 2, 3]
list2 = [4, 5]
combined = list1 + list2
print(combined)
# Expected output: [1, 2, 3, 4, 5]
````

````python
# Example 2: Repeat a list
print([0] * 5)
print(['hi'] * 3)
# Expected output:
# [0, 0, 0, 0, 0]
# ['hi', 'hi', 'hi']
````

## 1.18 copy(): Make a Real Copy

Writing `b = a` does **not** copy a list. It only gives the same list a second name, so changing `b` also changes `a`. Use `copy()` to get a separate, independent list.

**Syntax:** `new_list = my_list.copy()`

````python
# Example 1: The problem with b = a
a = [1, 2, 3]
b = a          # b is just another name for the SAME list
b.append(4)
print(a)
# Expected output: [1, 2, 3, 4]   (a changed too!)
````

````python
# Example 2: copy() makes a separate list
a = [1, 2, 3]
b = a.copy()
b.append(4)
print(a)
print(b)
# Expected output:
# [1, 2, 3]
# [1, 2, 3, 4]
````

## 1.19 Looping Through a List

A `for` loop visits each item one at a time, so you can do something with every item. Use `enumerate()` when you also need each item's position.

**Syntax:**
```python
for item in my_list:
    # do something with item
```

````python
# Example 1: Print each item
fruits = ['apple', 'banana', 'mango']
for fruit in fruits:
    print(fruit)
# Expected output:
# apple
# banana
# mango
````

````python
# Example 2: Print each item with its position
fruits = ['apple', 'banana', 'mango']
for position, fruit in enumerate(fruits):
    print(position, fruit)
# Expected output:
# 0 apple
# 1 banana
# 2 mango
````

````python
# Example 3: Build a new list from an old one (give a discount of 10 on each price)
prices = [100, 250, 80]
discounted = []
for p in prices:
    discounted.append(p - 10)
print(discounted)
# Expected output: [90, 240, 70]
````

## 1.20 Nested Lists (Lists Inside Lists)

A list can hold other lists. This is useful for table-like data with rows and columns.
- `table[row]` gives a whole row
- `table[row][column]` gives one value

```python
students = [['Asha', 85],    # row 0
            ['Ravi', 92]]    # row 1
```

````python
# Example 1: Get a row, then a single value
students = [['Asha', 85], ['Ravi', 92], ['Meena', 78]]
print(students[1])       # second row
print(students[1][0])    # second row, first column
print(students[2][1])    # third row, second column
# Expected output:
# ['Ravi', 92]
# Ravi
# 78
````

````python
# Example 2: Slicing inside a nested list
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(matrix[0][1:])   # first row, from index 1 to the end
print(matrix[:2])      # first two rows
# Expected output:
# [2, 3]
# [[1, 2, 3], [4, 5, 6]]
````

````python
# Example 3: Loop through each row
students = [['Asha', 85], ['Ravi', 92], ['Meena', 78]]
for row in students:
    print(row[0], 'scored', row[1])
# Expected output:
# Asha scored 85
# Ravi scored 92
# Meena scored 78
````

### List Quick Reference

| Operation | What it does |
|---|---|
| `lst[i]` | Get the item at index i |
| `lst[a:b]` | Get items from index a up to (not including) b |
| `append(x)` | Add x at the end |
| `extend(items)` | Add many items at the end |
| `insert(i, x)` | Add x at index i |
| `pop()` / `pop(i)` | Remove and return the last item / the item at i |
| `remove(x)` | Remove the first x |
| `del lst[i]` | Delete the item at index i |
| `clear()` | Remove everything |
| `index(x)` | Position of the first x |
| `count(x)` | How many times x appears |
| `sort()` | Sort the list itself |
| `sorted(lst)` | Return a new sorted list |
| `reverse()` | Flip the order |
| `copy()` | Make an independent copy |

# Part 2: Tuples

A **tuple** is like a list that is **locked**. Once created, you cannot add, remove or change its items. Use tuples for data that should never change, such as the days of the week or a location's latitude and longitude.

- Written with **round brackets** `( )`
- **Ordered:** items keep their positions, so indexing and slicing work exactly like lists
- **Unchangeable (immutable):** no append, insert, remove or update
- **Allows duplicates**
- Slightly faster and lighter than lists, and safe from accidental changes
- Can be used as dictionary keys (lists cannot)

## 2.1 Creating a Tuple

Put items inside round brackets, separated by commas.

**Watch out:** a tuple with only one item needs a comma after it: `(5,)`. Without the comma, `(5)` is just the number 5 in brackets.

**Syntax:**
```python
my_tuple = (item1, item2, item3)
```

````python
# Example 1: A tuple of text and a tuple of numbers
days = ('Mon', 'Tue', 'Wed')
point = (10, 20)
print(days)
print(point)
# Expected output:
# ('Mon', 'Tue', 'Wed')
# (10, 20)
````

````python
# Example 2: A tuple with ONE item needs a comma
not_a_tuple = (5)
real_tuple = (5,)
print(type(not_a_tuple))
print(type(real_tuple))
# Expected output:
# <class 'int'>
# <class 'tuple'>
````

````python
# Example 3: Brackets are optional, and tuple() converts a list into a tuple
colors = 'red', 'green', 'blue'
nums = tuple([1, 2, 3])
print(colors)
print(nums)
# Expected output:
# ('red', 'green', 'blue')
# (1, 2, 3)
````

## 2.2 Indexing and Slicing

Tuples use positions exactly like lists: indexes start at 0, negative indexes count from the end, and slicing uses `[start:stop:step]`. Slicing a tuple gives you a new tuple.

````python
# Example 1: Indexing
days = ('Mon', 'Tue', 'Wed', 'Thu', 'Fri')
print(days[0])
print(days[-1])
# Expected output:
# Mon
# Fri
````

````python
# Example 2: Slicing
days = ('Mon', 'Tue', 'Wed', 'Thu', 'Fri')
print(days[1:3])
print(days[::-1])
# Expected output:
# ('Tue', 'Wed')
# ('Fri', 'Thu', 'Wed', 'Tue', 'Mon')
````

## 2.3 Tuples Cannot Be Changed (Immutability)

Trying to change, add or remove an item in a tuple gives a `TypeError`. If you really need a change, convert the tuple to a list, change the list, and convert it back.

````python
# Example 1: Trying to change an item
days = ('Mon', 'Tue', 'Wed')
# Remove the # from the next line to see the error:
# days[0] = 'Sun'
# Expected output (after removing #): TypeError: 'tuple' object does not support item assignment
````

````python
# Example 2: The workaround: tuple -> list -> change -> tuple
days = ('Mon', 'Tue', 'Wed')
temp = list(days)
temp[0] = 'Sun'
days = tuple(temp)
print(days)
# Expected output: ('Sun', 'Tue', 'Wed')
````

## 2.4 count() and index()

Because tuples cannot change, they have only two methods:
- `count(item)`: how many times the item appears
- `index(item)`: the position of the first match

````python
# Example 1: Count repeated values
ratings = (5, 4, 5, 3, 5)
print(ratings.count(5))
# Expected output: 3
````

````python
# Example 2: Find the position of an item
days = ('Mon', 'Tue', 'Wed')
print(days.index('Wed'))
# Expected output: 2
````

## 2.5 Unpacking: Split a Tuple into Variables

Unpacking takes each item out of a tuple and stores it in its own variable, all in one line. The number of variables must match the number of items.

**Syntax:** `a, b, c = my_tuple`

````python
# Example 1: Unpack a person's details
person = ('Asha', 21, 'Delhi')
name, age, city = person
print(name)
print(age)
print(city)
# Expected output:
# Asha
# 21
# Delhi
````

````python
# Example 2: Swap two values in one line (this uses tuples behind the scenes)
a = 5
b = 10
a, b = b, a
print(a, b)
# Expected output: 10 5
````

## 2.6 Other Useful Operations

`len()`, `in`, `+`, `*`, `min()`, `max()` and `for` loops all work on tuples, just like on lists.

````python
# Example 1: len, in, min and max
marks = (78, 92, 85)
print(len(marks))
print(92 in marks)
print(min(marks), max(marks))
# Expected output:
# 3
# True
# 78 92
````

````python
# Example 2: Join, repeat and loop
t = (1, 2) + (3, 4)
print(t)
print(('ha',) * 3)
for item in t:
    print(item)
# Expected output:
# (1, 2, 3, 4)
# ('ha', 'ha', 'ha')
# 1
# 2
# 3
# 4
````

### Tuple Quick Reference

| Operation | What it does |
|---|---|
| `t[i]`, `t[a:b]` | Index and slice, same as lists |
| `count(x)` | How many times x appears |
| `index(x)` | Position of the first x |
| `a, b = t` | Unpack into variables |
| `list(t)` / `tuple(lst)` | Convert between tuple and list |

**List or tuple?** If the data will change, use a list. If it should stay fixed, use a tuple.

# Part 3: Sets

A **set** is like a bag of **unique** items. It automatically throws away duplicates and has **no order**, so there are no positions (no indexing or slicing). Sets are great for removing duplicates and for comparing groups: what is common, and what is different.

- Written with **curly brackets** `{ }`
- **No duplicates:** each value is stored only once
- **Unordered:** items have no index numbers
- **Changeable:** you can add and remove items
- Items must be unchangeable types (numbers, text, tuples); a list cannot go inside a set

> **Note:** Because sets have no order, Python may print items in a different order from the Expected output. The items themselves will be the same.

## 3.1 Creating a Set

Put items inside curly brackets, or convert a list with `set()`.

**Watch out:** an empty set must be created with `set()`. Writing `{}` creates an empty **dictionary**, not a set.

**Syntax:**
```python
my_set = {item1, item2, item3}
```

````python
# Example 1: Duplicates are removed automatically
nums = {1, 2, 2, 3, 3, 3}
print(nums)
# Expected output: {1, 2, 3}
````

````python
# Example 2: Remove duplicates from a list using set()
roll_numbers = [3, 1, 3, 2, 1]
unique = set(roll_numbers)
print(unique)
print(len(roll_numbers), len(unique))
# Expected output:
# {1, 2, 3}
# 5 3
````

````python
# Example 3: An EMPTY set needs set(), because {} makes an empty dictionary
empty_set = set()
empty_dict = {}
print(type(empty_set))
print(type(empty_dict))
# Expected output:
# <class 'set'>
# <class 'dict'>
````

## 3.2 No Indexing in Sets

## 3.3 add() and update(): Adding Items

- `add(item)` adds **one** item. Adding something that is already there does nothing.
- `update(items)` adds **many** items from a list, tuple or another set.

**Syntax:** `my_set.add(item)` and `my_set.update(collection)`

````python
# Example 1: Add one item
nums = {1, 2, 3}
nums.add(4)
nums.add(2)    # 2 is already there, so nothing changes
print(nums)
# Expected output: {1, 2, 3, 4}
````

````python
# Example 2: Add many items at once
nums = {1, 2}
nums.update([3, 4, 5])
print(nums)
# Expected output: {1, 2, 3, 4, 5}
````

````python
# Example 3: Adding text items
skills = {'python', 'sql'}
skills.add('excel')
skills.add('python')   # duplicate, ignored
print(len(skills))
print('excel' in skills)
# Expected output:
# 3
# True
````

## 3.4 remove(), discard(), pop() and clear(): Removing Items

- `remove(item)` deletes the item, but gives a `KeyError` if it is not there
- `discard(item)` deletes the item and **stays quiet** if it is not there (the safer choice)
- `pop()` removes and returns **any one** item. You cannot choose which, because sets have no order
- `clear()` empties the whole set

````python
# Example 1: remove() vs discard()
nums = {1, 2, 3}
nums.remove(2)
nums.discard(10)   # 10 is not in the set, but there is no error
print(nums)
# nums.remove(10)  # Remove the # at the start to see: KeyError: 10
# Expected output: {1, 3}
````

````python
# Example 2: pop() and clear()
nums = {1, 2, 3}
item = nums.pop()
print(item)        # could be any item from the set
print(len(nums))
nums.clear()
print(nums)
# Expected output:
# 1   (you may see a different item)
# 2
# set()
````

## 3.5 Set Operations: Comparing Two Groups

This is where sets really shine. Imagine two groups of students, listed by roll number: those who play cricket and those who play football.

- **Union** `a | b` or `a.union(b)`: everyone in **either** group (no duplicates)
- **Intersection** `a & b` or `a.intersection(b)`: only those in **both** groups
- **Difference** `a - b` or `a.difference(b)`: those in `a` but **not** in `b`
- **Symmetric difference** `a ^ b` or `a.symmetric_difference(b)`: those in **exactly one** group

Each of these gives you a **new** set; the original sets do not change.

````python
# Example 1: Union - students who play at least one game
cricket = {1, 2, 3, 4}     # roll numbers of cricket players
football = {3, 4, 5, 6}    # roll numbers of football players
print(cricket | football)
print(cricket.union(football))
# Expected output:
# {1, 2, 3, 4, 5, 6}
# {1, 2, 3, 4, 5, 6}
````

````python
# Example 2: Intersection - students who play BOTH games
cricket = {1, 2, 3, 4}
football = {3, 4, 5, 6}
print(cricket & football)
print(cricket.intersection(football))
# Expected output:
# {3, 4}
# {3, 4}
````

````python
# Example 3: Difference - plays one game but NOT the other
cricket = {1, 2, 3, 4}
football = {3, 4, 5, 6}
print(cricket - football)   # cricket only
print(football - cricket)   # football only
# Expected output:
# {1, 2}
# {5, 6}
````

````python
# Example 4: Symmetric difference - plays exactly ONE of the two games
cricket = {1, 2, 3, 4}
football = {3, 4, 5, 6}
print(cricket ^ football)
print(cricket.symmetric_difference(football))
# Expected output:
# {1, 2, 5, 6}
# {1, 2, 5, 6}
````

## 3.6 Subset, Superset and Disjoint

- `a.issubset(b)`: is every item of `a` also in `b`?
- `a.issuperset(b)`: does `a` contain every item of `b`?
- `a.isdisjoint(b)`: do `a` and `b` have **nothing** in common?

All three answer with `True` or `False`.

````python
# Example 1: Does the candidate have all the required skills?
required = {'python', 'sql'}
candidate = {'python', 'sql', 'excel'}
print(required.issubset(candidate))
print(candidate.issuperset(required))
# Expected output:
# True
# True
````

````python
# Example 2: Do two batches have no students in common?
morning_batch = {'Asha', 'Ravi'}
evening_batch = {'Meena', 'John'}
print(morning_batch.isdisjoint(evening_batch))
# Expected output: True
````

## 3.7 Looping, len() and Sorting a Set

A set has no order, but you can still loop through it, count it with `len()`, or get a sorted **list** from it with `sorted()` when you need a fixed order.

````python
# Example 1: Unique values in a fixed order, using set() and sorted()
cities = ['Delhi', 'Mumbai', 'Delhi', 'Chennai', 'Mumbai']
unique_cities = sorted(set(cities))
print(unique_cities)
print(len(unique_cities))
# Expected output:
# ['Chennai', 'Delhi', 'Mumbai']
# 3
````

````python
# Example 2: Loop through a set in alphabetical order
tags = {'python', 'beginner', 'lab'}
for tag in sorted(tags):
    print(tag)
# Expected output:
# beginner
# lab
# python
````

# Part 4: Dictionaries

A **dictionary** works like a contact book: you look up a **name (the key)** to find a **phone number (the value)**. Instead of positions like 0, 1, 2, you use meaningful keys.

Each entry is a **key: value** pair. In `{'name': 'Asha'}`, `'name'` is the key and `'Asha'` is the value.

- Written with **curly brackets** `{ }` containing `key: value` pairs
- **Keys must be unique:** using the same key again replaces the old value
- **Keys must be unchangeable types:** text, numbers or tuples (not lists)
- **Values can be anything:** numbers, text, lists, even other dictionaries
- **Ordered:** since Python 3.7, items stay in the order you added them
- **Changeable:** you can add, update and delete pairs

## 4.1 Creating a Dictionary

**Syntax:**
```python
my_dict = {key1: value1, key2: value2}
```

````python
# Example 1: A dictionary with key: value pairs
student = {'name': 'Asha', 'age': 21, 'city': 'Delhi'}
print(student)
# Expected output: {'name': 'Asha', 'age': 21, 'city': 'Delhi'}
````

````python
# Example 2: Using dict(), and an empty dictionary
car = dict(brand='Tata', model='Nexon', year=2024)
empty = {}
print(car)
print(empty)
# Expected output:
# {'brand': 'Tata', 'model': 'Nexon', 'year': 2024}
# {}
````

````python
# Example 3: Values can be lists, and a repeated key keeps only the LAST value
student = {'name': 'Ravi', 'marks': [85, 90, 78], 'name': 'Ravi Kumar'}
print(student)
# Expected output: {'name': 'Ravi Kumar', 'marks': [85, 90, 78]}
````

## 4.2 Accessing Values

- `my_dict[key]` gets the value, but gives a `KeyError` if the key does not exist
- `my_dict.get(key)` gets the value, and gives `None` (or a default you choose) if the key does not exist. **This is the safer way.**

**Syntax:** `my_dict[key]` or `my_dict.get(key, default)`

````python
# Example 1: Access with square brackets
student = {'name': 'Asha', 'age': 21, 'city': 'Delhi'}
print(student['name'])
print(student['age'])
# Expected output:
# Asha
# 21
````

````python
# Example 2: Access safely with get()
student = {'name': 'Asha', 'age': 21}
print(student.get('city'))                # key missing -> None
print(student.get('city', 'Not given'))   # key missing -> your default
# print(student['city'])  # Remove the # at the start to see: KeyError: 'city'
# Expected output:
# None
# Not given
````

## 4.3 Adding and Updating Items

Use `my_dict[key] = value`:
- If the key **already exists**, its value is **updated**
- If the key is **new**, the pair is **added**

**Syntax:** `my_dict[key] = value`

````python
# Example 1: Update an existing key
student = {'name': 'Asha', 'age': 21}
student['age'] = 22
print(student)
# Expected output: {'name': 'Asha', 'age': 22}
````

````python
# Example 2: Add a new key
student = {'name': 'Asha', 'age': 21}
student['city'] = 'Delhi'
print(student)
# Expected output: {'name': 'Asha', 'age': 21, 'city': 'Delhi'}
````

````python
# Example 3: Build a dictionary step by step (count the fruits in a basket)
basket = ['apple', 'mango', 'apple', 'apple']
counts = {}
for fruit in basket:
    counts[fruit] = counts.get(fruit, 0) + 1   # start at 0 if the fruit is new
print(counts)
# Expected output: {'apple': 3, 'mango': 1}
````

## 4.4 update(): Add or Change Many Items at Once

`update()` merges another dictionary into this one. Keys that already exist get the new values; new keys are added.

**Syntax:** `my_dict.update(other_dict)`

````python
# Example 1: Update one key and add another in one step
student = {'name': 'Asha', 'age': 21}
student.update({'age': 22, 'city': 'Delhi'})
print(student)
# Expected output: {'name': 'Asha', 'age': 22, 'city': 'Delhi'}
````

````python
# Example 2: Apply a user's choices on top of default settings
settings = {'theme': 'light', 'font': 12}
user_choice = {'theme': 'dark'}
settings.update(user_choice)
print(settings)
# Expected output: {'theme': 'dark', 'font': 12}
````

## 4.5 Removing Items: pop(), popitem(), del and clear()

- `pop(key)` removes the pair and **gives back the value**
- `pop(key, default)` returns the default instead of an error if the key is missing
- `popitem()` removes and returns the **last added** pair as a tuple
- `del my_dict[key]` deletes the pair (gives nothing back)
- `clear()` empties the whole dictionary

````python
# Example 1: pop() removes a key and gives back its value
student = {'name': 'Asha', 'age': 21, 'city': 'Delhi'}
age = student.pop('age')
print(age)
print(student)
# Expected output:
# 21
# {'name': 'Asha', 'city': 'Delhi'}
````

````python
# Example 2: pop() with a default avoids an error for a missing key
student = {'name': 'Asha'}
print(student.pop('phone', 'No phone saved'))
# Expected output: No phone saved
````

````python
# Example 3: popitem(), del and clear()
student = {'name': 'Asha', 'age': 21, 'city': 'Delhi'}
last = student.popitem()
print(last)
del student['age']
print(student)
student.clear()
print(student)
# Expected output:
# ('city', 'Delhi')
# {'name': 'Asha'}
# {}
````

## 4.6 keys(), values() and items()

- `keys()` gives all the keys
- `values()` gives all the values
- `items()` gives all the pairs as `(key, value)` tuples

These return special view objects. Wrap them in `list()` if you need a normal list.

````python
# Example 1: See the keys, values and items
student = {'name': 'Asha', 'age': 21}
print(student.keys())
print(student.values())
print(student.items())
# Expected output:
# dict_keys(['name', 'age'])
# dict_values(['Asha', 21])
# dict_items([('name', 'Asha'), ('age', 21)])
````

````python
# Example 2: Convert the keys into a normal list
student = {'name': 'Asha', 'age': 21}
key_list = list(student.keys())
print(key_list)
print(key_list[0])
# Expected output:
# ['name', 'age']
# name
````

## 4.7 Checking Keys with in, and Counting with len()

`in` checks the **keys**, not the values. To search the values, use `in my_dict.values()`. `len()` counts the key: value pairs.

````python
# Example 1: 'in' looks at keys
student = {'name': 'Asha', 'age': 21}
print('name' in student)
print('Asha' in student)            # 'Asha' is a value, not a key
print('Asha' in student.values())
print(len(student))                 # number of pairs
# Expected output:
# True
# False
# True
# 2
````

````python
# Example 2: Check before accessing, to avoid a KeyError
prices = {'pen': 10, 'notebook': 50}
item = 'bag'
if item in prices:
    print(prices[item])
else:
    print(item, 'is not available')
# Expected output: bag is not available
````

## 4.8 Looping Through a Dictionary

- `for key in my_dict:` loops through the keys
- `for key, value in my_dict.items():` gives you both at once (the most common way)

````python
# Example 1: Loop through the keys
prices = {'pen': 10, 'notebook': 50, 'bag': 500}
for item in prices:
    print(item)
# Expected output:
# pen
# notebook
# bag
````

````python
# Example 2: Loop through keys AND values with items()
prices = {'pen': 10, 'notebook': 50, 'bag': 500}
for item, price in prices.items():
    print(item, 'costs', price)
# Expected output:
# pen costs 10
# notebook costs 50
# bag costs 500
````

````python
# Example 3: Add up all the values
prices = {'pen': 10, 'notebook': 50, 'bag': 500}
print(sum(prices.values()))
# Expected output: 560
````

## 4.9 Nested Dictionaries

A value can itself be a dictionary, which is handy for storing records, such as one record per student. Use one pair of square brackets for each level: `students['S2']['name']`.

````python
# Example 1: A dictionary of students (each value is another dictionary)
students = {
    'S1': {'name': 'Asha', 'marks': 85},
    'S2': {'name': 'Ravi', 'marks': 92}
}
print(students['S2'])
print(students['S2']['name'])
# Expected output:
# {'name': 'Ravi', 'marks': 92}
# Ravi
````

````python
# Example 2: Update a nested value, then loop through all records
students = {
    'S1': {'name': 'Asha', 'marks': 85},
    'S2': {'name': 'Ravi', 'marks': 92}
}
students['S1']['marks'] = 88
for roll, info in students.items():
    print(roll, info['name'], info['marks'])
# Expected output:
# S1 Asha 88
# S2 Ravi 92
````

## 4.10 Converting a Dictionary to JSON Text

To save a dictionary to a file or send it to a website or API, convert it to **JSON** text with `json.dumps()`. Use `json.loads()` to turn JSON text back into a dictionary.

**Watch out:** `str(my_dict)` also gives text, but with single quotes. That is Python style and **not** valid JSON.

````python
# Example 1: Dictionary -> JSON text
import json
address = {'street': '180 Adams Street', 'city': 'Chicago', 'zip': '60611'}
print(json.dumps(address))   # valid JSON (double quotes)
print(str(address))          # Python style (single quotes), NOT valid JSON
# Expected output:
# {"street": "180 Adams Street", "city": "Chicago", "zip": "60611"}
# {'street': '180 Adams Street', 'city': 'Chicago', 'zip': '60611'}
````

````python
# Example 2: JSON text -> dictionary
import json
text = '{"name": "Asha", "age": 21}'
data = json.loads(text)
print(data['name'])
print(type(data))
# Expected output:
# Asha
# <class 'dict'>
````

# Part 5: Quick Recap

### Which one should I use?
- An ordered collection you will keep changing → **List**
- Fixed data that must never change → **Tuple**
- Unique items only, or comparing two groups → **Set**
- Looking up a value by a name or ID → **Dictionary**

### Common beginner mistakes
1. **Indexes start at 0**, so the first item is `my_list[0]`, not `my_list[1]`.
2. **The stop value in a slice is not included:** `nums[1:3]` gives the items at index 1 and 2 only.
3. **`(5)` is not a tuple**, it is just the number 5. Write `(5,)`.
4. **`{}` is an empty dictionary**, not an empty set. Use `set()`.
5. **`sort()` returns nothing.** Writing `marks = marks.sort()` turns `marks` into `None`. Just write `marks.sort()`.
6. **`b = a` does not copy a list.** Use `b = a.copy()`.
7. **Never name a variable `list`, `tuple`, `set`, `dict` or `str`.** These names belong to Python, and reusing them breaks the built-in functions (for example, `list(...)` stops working).
